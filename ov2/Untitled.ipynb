```python
import pandas as pd

dfm = pd.read_csv("moodle2.csv")
dfm = dfm.replace("-", 0.0)
dfm = dfm.astype({"Quiz: Øving 2 - Funksjoner, løkker og lister (Real)": "float64",
           "Quiz: Øving 1 - Variabler og datatyper (Real)": "float64"})

dfm["pass"]  = (dfm["Quiz: Øving 1 - Variabler og datatyper (Real)"] >= 7.5) & (dfm["Quiz: Øving 2 - Funksjoner, løkker og lister (Real)"] >= 7.5)

dfm["Godkjent"] = dfm["pass"].map(lambda x: "Godkjent" if x else "Ikke godkjent")

dfm.query("`Last name` == 'Grønmo'")
#dfm.query("`First name`== 'Fredrik'")

```
