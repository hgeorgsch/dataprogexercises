# Avfall
* SSB tabell 14458: Industriavfall. Generert mengde avfall (1 000 tonn), etter materialtype, statistikkvariabel, år og behandlingsmåte
* Alle statistikkvariabler



```python
import pandas as pd
import numpy as np

df = pd.read_csv("SSB_14458_Avfall.csv", sep="\t", encoding="latin1")
display(df)

df=df.melt(id_vars=["materialtype", "behandlingsmåte"], 
        value_vars=["Avfallsmengde 2015", "Avfallsmengde 2022"], 
        var_name="År", 
        value_name="tonn")
df["År"] = df["År"].map(lambda streng: int(streng[-4:]))
display(df)

df.pivot(columns="År", values="tonn", index=["materialtype", "behandlingsmåte"])
```
