```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

n = 180
populasjon = np.random.normal(10.0,0.1,n) # Trekker tilfeldige tall med normalfordeling rundt 10
arbeidsledighet = np.random.normal(0.03,0.002,n) # Trekker tilfeldig arbeidsledighet, normalfordelt

indeks = [f"{year}M{month:02d}"
              for year in range(1990, 2005)
              for month in range(1,13)
         ]
arbeidsledighet_formatert = [f"{ledighet:.2%}" for ledighet in arbeidsledighet]
df = pd.DataFrame({"Populasjon (million)": populasjon, "Arbeidsledighet": arbeidsledighet_formatert}, index=indeks)

df["arbeidsledighet_decimal"] = arbeidsledighet
df["Antall uten arbeid(tusen)"] = df["arbeidsledighet_decimal"]*df["Populasjon (million)"]*1000
df["Antall uten arbeid(tusen)"] = df["Antall uten arbeid(tusen)"].apply(lambda tall: int(round(tall,0)))
df = df.drop(columns="arbeidsledighet_decimal")
display(df)

#Alternativ med pandas.Period som indeks
indeksPeriode = pd.period_range('1990', '2005', freq='M') #Lag periodeindeks
df2 = df.set_index(indeksPeriode[:-1]) # Sett ny indeks -- indeksPeriode går til og med 2005-01, derfor slicer vi
df2
```

```python
import datetime


def month_to_quarter(df, kumulative=[]):
    """
    df: dataframe med indeks på månedsform yyyyMmm (ie 1987M04 = april 1987)
    
    kumulative: En liste med kolonnenavn hvor kvartalsverdien 
        skal være summen av månedsverdiene. Default er en tom liste.
        For øvrige kolonner blir kvartalsverdien gjennomsnittsverdien av månedsverdiene

    Funksjonen returnerer et nytt dataframe med indeks på kvartalsform yyyyKq (ie 1987K2 for andre kvartal 1987)
    """
    # Kode Kode
    dato = df.index.map(lambda x: datetime.datetime.strptime(str(x), "%YM%m"))
    kvartal = dato.to_period('Q')
    kvartal.name = "kvartal"
    df2 = df.set_index(kvartal)
    group = df2.groupby("kvartal")
    df_new = pd.DataFrame(index=df2.index)
    for col in df.columns:
        if col in kumulative and df[col].dtype not in ['object', 'string']:
            df_new[col] = group[col].sum()
        elif df[col].dtype not in ['object', 'string']:
            df_new[col] = group[col].mean()
    
    return df_new.drop_duplicates()
```

```python
def quarter_to_year(df, kumulative=[]):
    """
    df: dataframe med indeks på kvartalsform yyyyKk (ie 1987K2 = andre kvartal 1987)
    
    kumulative: En liste med kolonnenavn hvor årsverdien 
        skal være summen av kvartalsverdiene. Default er en tom liste.
        For øvrige kolonner blir årsverdien gjennomsnittsverdien av kvartalsverdiene

    Funksjonen returnerer et nytt dataframe med indeks på årsform yyyy (ie 1987 for herrens år 1987)
    """
    # Kode Kode
    dato = df.index.map(lambda x: datetime.datetime.strptime(str(x), "%YQ%S")) #Les til datetime
    year = dato.to_period('Y') #Gjør om til pandas periode
    year.name = "year" #sett navn
    df2 = df.set_index(year) #lag ny dataframe med nye year indeks
    group = df2.groupby("year") #Grupper alle år kvartal i året
    df_new = pd.DataFrame(index=df2.index) 
    for col in df.columns:
        if col in kumulative and df[col].dtype not in ['object', 'string']:
            df_new[col] = group[col].sum()
        elif df[col].dtype not in ['object', 'string']:
            df_new[col] = group[col].mean()
    
    return df_new.drop_duplicates()
    
```
