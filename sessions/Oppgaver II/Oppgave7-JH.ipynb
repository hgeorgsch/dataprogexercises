```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

boforhold_df = pd.read_csv("boforhold2.csv", encoding="ISO-8859-1", sep=";", header=1)
boforhold_df = boforhold_df.melt(id_vars=["inntektsgruppe", "statistikkvariabel"], var_name="År", value_name="prosent")
boforhold_df.drop("statistikkvariabel", axis=1, inplace=True)
#boforhold_df = boforhold_df.melt(id_vars=["statistikkvariabel", "inntektsgruppe"], var_name="År", value_name="prosent")
boforhold_df = boforhold_df.pivot(columns="inntektsgruppe", index="År")
boforhold_df.index = pd.PeriodIndex(boforhold_df.index, freq='Y')
boforhold_df.plot()
```
