<h1 style="font-size:3rem;color:blue;">Populasjonsvekst</h1>

```python
import matplotlib.pyplot as plt
import numpy as np

def beregn_populasjon(P0, K, r, tid):
    e = 2.71828
    A = (K - P0) / P0
    populasjon = K / (1 + A * e**(-r * tid))
    return populasjon

def plot_populasjon(P0, K, r, start, slutt):
    tid = np.arange(start, slutt + 1)
    populasjon = [beregn_populasjon(P0, K, r, t - start) for t in tid]

    plt.plot(tid, populasjon, label='Populasjon')
    plt.title('Populasjonsutvikling over tid')
    plt.xlabel('År')
    plt.ylabel('Populasjon')
    plt.legend()
    plt.grid(True)
    plt.show()

# Parameteret for oppgaven
K = 11e6
r = 0.02
start = 2023
slutt = 2043
P0 = 10521556

print("Ved år", start, "var innbyggertallet på Sverige", P0)
print("Vi antar en årlig vekstrate på", r*100, "%")


# Funksjonen og plot av resultatet
populasjon_2043 = beregn_populasjon(P0, K, r, slutt - start)
print("Ved år", slutt, "vil populasjonen være omtrent", int(populasjon_2043), "\n \n")

# Plot av populasjonen over tid
plot_populasjon(P0, K, r, start, slutt)
```
