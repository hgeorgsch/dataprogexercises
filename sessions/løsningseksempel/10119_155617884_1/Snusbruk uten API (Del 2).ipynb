Problemstilling: En undersøkelse av endringer i snusingens utbredelse fra 2018 til 2022: Hvorvidt andelen mennesker som snuser har økt i aldersgruppen unge voksne (16-24 år)

I dette Programmet vil jeg legge inn data hentet fra SSB (Statistisk Sentralbyrå) fra tabell: "07692: Daglig snusbrukere og av-og-til-snusbrukere (prosent), etter kjønn, alder, statistikkvariabel og år" Med spesielt fokus på daglig snusbruk i aldresgruppen 16 - 24 år.

```python
# Import av dataanalyse fra SSB
# Daglig snusbruk blandt unge fra 16 - 24 år fra 2018 til 2022

import pandas as pd
import matplotlib.pyplot as plt

# Mitt datasett
data = {
    'År': [2018, 2019, 2020, 2021, 2022],
    'Snus_Bruk_Prosent': [19, 20, 18, 21, 22]
}

# DataFrame fra datasettet
dataset = pd.DataFrame(data)


# Plott av snusbruk over tid
plt.plot(dataset['År'], dataset['Snus_Bruk_Prosent'])
plt.xlabel('År')
plt.ylabel('Andel som bruker snus (%)')
plt.title('Snusbruk blant unge fra 16-24 år over tid')
plt.show()
```

Konklusjon: Ut i fra dette programmet ble det tydelig at andelen unge som bruker snus daglig har økt. Og det hadde en betydelig etter en nedgang i 2020, man kan spørre seg hva som kan være årsaken til dette. Kan foreksempel pandemien være en faktor som spiller inn? 

Dette bekrefter min misstanke om at andelen unge som snuser har økt.
