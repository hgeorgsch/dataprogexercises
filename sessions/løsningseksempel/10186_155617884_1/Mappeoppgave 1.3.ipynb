## Lånekalkulator - Annuitetslån

### Lånekalkulator 1: Hvor mye kan du låne?

```python
def boliglån_test():
    EK = int(input("Hvor stor egenkapital kan du gå inn med? \n"))
    GJ = int(input("Hva er din samlede gjeld utenom evt. gjeld som innløses ved opptak av ny gjeld? \n"))
    Lønn = int(input("Hva er din brutto årsinntekt? \n"))

    EK_1 = EK-GJ
    Boliglån_max = EK_1/15*100
    Boliglån_max_2 = Lønn*5
    
    if GJ>EK:
        print("Du har negativ egenkapital, som betyr at du dessverre ikke kan tilfredsstille kravet om egenkapital for å få boliglån. \n")
    
    else:
        if Boliglån_max_2>Boliglån_max:
            print("Du kan låne", round(Boliglån_max),". Du kan låne inntil 5 ganger din brutto årsinntekt, og du trenger minimum 15% egenkapital. \n")
        
        else:
            print("Du kan låne", round(Boliglån_max_2),". Du kan låne inntil 5 ganger din brutto årsinntekt, og du trenger minimum 15% egenkapital. \n")
        
boliglån_test()
```

### Lånekalkulator 2: Hvor stor egenkapital trenger du?

```python
Pris = int(input("Hvor mye koster boligen inklusive evt. fellesgjeld? \n"))

EK_2 = round(Pris/100*15)

print("Du trenger en egenkapital på minimum", EK_2,".", "Dette fordi du må ha minimum 15% egenkapital av totalprisen på boligen.")
```

### Lånekalkulator 3: Hva koster lånet?

```python
Lån = float(input("Hvor mye skal du låne? \n"))
Tid = float(input("Hvor lang nedbetalingstid planlegger du å ha på ditt lån? Vanlig nedbetalingstid på boliglån ligger typisk mellom 20 og 30 år. \n"))
Rente = float(input("Til hvilken effektiv rente per år skal du ta opp lån? Oppgi rente i prosent uten prosenttegn. \n"))
Termingebyr = float(input("Hva er ditt månedlige termingebyr? \n"))
Etableringsgebyr = float(input("Hva er beløpet for etableringsgebyret for ditt lån? \n"))

Annuitetsbeløp_måned = Lån * (Rente/100/12*(1+Rente/100/12)**(Tid*12)) / ((1 + Rente/100/12)**(Tid*12)-1)
Totalbeløp = Annuitetsbeløp_måned * 12 * 25 + Etableringsgebyr + Termingebyr * 12 * 25
Totalkostnad = Totalbeløp - Lån


print("Ditt totalbeløp er:", round(Totalbeløp, 2), "\n Din totale rentekostnad inklusive gebyrer er:", round(Totalkostnad, 2), "\n Ditt månedlige annuitetsbeløp er:", round(Annuitetsbeløp_måned, 2))
```
