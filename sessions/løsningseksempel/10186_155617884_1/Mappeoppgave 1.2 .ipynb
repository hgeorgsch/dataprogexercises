## Sparekalkulator

Denne sparekalkulatoren tar hensyn til startbeløp fra og med 0, og med eller uten terminsparing. Dersom man svarer 0 terminer på spørsmålet om antall terminer per år, vil den ikke spørre etter terminbeløp pga. if setningen som tilhører "if terminer>0:".

```python
startbeløp = int(input("Hva er ditt startbeløp? \n"))
terminer = int(input("Hvor mange terminer per år ønsker du? \n"))
if terminer>0:
    terminbeløp = int(input("Hvor mye ønsker du å spare per termin? \n"))
rente = float(input("Hvilken rente forventer du å få? Oppgi tall i prosent uten prosenttegn. \n"))/100
år = int(input("Hvor mange år ønsker du å spare? \n"))

def sparekalkulator():
    if terminer>0:
        saldo = round(terminbeløp*(((1+rente/terminer)**(terminer*år)-1)/(rente/terminer))+(startbeløp*(rente+1)**år))
        print("Med informasjonen du har oppgitt, vil din saldo etter", år, "år være", saldo, "kroner.")
        
    else:
        saldo = round(startbeløp*(1+rente)**år)
        print("Med informasjonen du har oppgitt, vil din saldo etter", år, "år være", saldo, "kroner.")
sparekalkulator()
```
