```python
import matplotlib.pyplot as plt
import numpy as np

# Min sparekalkulator
P = float(input("Skriv inn ønsket månedsbeløp:"))
r = float(input("Skriv inn tilnærmet årlig rente (i prosent %):")) / 100
n = int(input("Skriv inn forventet antall innbetalinger i løpet av ett år:"))
t = int(input("Skriv inn antall år:"))

# Formelen for å beregne sluttbeløpet
def calculate_end_amount(P, r, n, t):
    return P * ((1 + r/n)**(n*t) - 1) / (r/n)

sluttbelop = calculate_end_amount(P, r, n, t)

print(f"Vekst på din sparekonto over gitt periode: {sluttbelop:.2f} kroner")

# Input for å lage en graf som viser endringen
def plot_savings_growth(P, r, n, t):
    periods = np.arange(0, t+1, 1)
    savings = [P * ((1 + r/n)**(n*i) - 1) / (r/n) for i in periods]

    plt.plot(periods, savings, marker='o', linestyle='-')
    plt.title('Din Sparekonto sin vekst over Tid')
    plt.xlabel('År')
    plt.ylabel('Beløp (i kroner)')
    plt.grid(True)
    plt.show()

# Plot for grafen
plot_savings_growth(P, r, n, t)
```
