```python
lånebeløp = float(input("Lånebeløp: "))
årlig_rentesats = float(input("Årlig rentesats - oppgi i hele tall, mener du renten er 3 prosent skriver du 3: ")) / 100
antall_år = int(input("Antall år: "))

månedlig_rentesats = årlig_rentesats / 12
antall_måneder = antall_år * 12

månedlig_avdrag = lånebeløp / antall_måneder

print("Måned   Månedlig Avdrag   Gjenstående Lånebeløp")
for måned in range(1, antall_måneder + 1):
    gjenværende_lån = lånebeløp - måned * månedlig_avdrag
    print(f"{måned:6}   {månedlig_avdrag:15.2f}   {gjenværende_lån:20.2f}")

total_kostnad = lånebeløp + lånebeløp * årlig_rentesats * antall_år
print(f"\nTotal kostnad for lånet: {total_kostnad:.2f} kroner")

#Dette er mye likt som i spareklakulator, men i tillegg har jeg bedt programmet om å gi en nedbetalingsplan ved hjelp av denne koden du ser i oppgaven.

#Jeg har også benyttet koden "<h1 style="font-size:3rem;color:blue;">Lånekalkulator // Håvard</h1>" for å lage en overskrift til hver oppgave.

```
