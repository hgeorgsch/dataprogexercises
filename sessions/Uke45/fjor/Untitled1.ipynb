Bruk web-api til https://opentdb.com/ til å lage et pythonprogram som kjører en quiz!

* Velg selv om du vil ha flere spillere spille alene.

```python
import requests
import json
import random
from html import unescape
    
kategori_url =  "https://opentdb.com/api_category.php"
quiz_url = "https://opentdb.com/api.php"

n = int(input("Hvor mange spørsmål vil du ha?\n"))

response = requests.get(kategori_url)
kategorier_data = response.json()
kategorier = kategorier_data["trivia_categories"]

print("Hvilken kategori?\n")
i=1
for kats in kategorier:
    print(f"({i}): {kats['name']}")
    i += 1
valg = int(input("\n"))

i=1
kat_id = -1
for kats in kategorier:
    if i == valg:
        kat_id = kats["id"]
    i += 1


parameters = {"amount": n, "category": kat_id, "difficulty": "medium", "type": "multiple"}    
response = requests.get(quiz_url, params=parameters)
question_data = response.json()

antall_riktige = 0
print("Flotters!\n Da begynner vi:")
i = 1
for q in question_data["results"]:
    print("Antall riktige svar:", antall_riktige)
    print(f"Spørsmål nummer {i}:")
    print(unescape(q["question"]))
    korrekt = q["correct_answer"]
    feil = q["incorrect_answers"]
    stokket = [korrekt]
    for alt in feil:
        stokket.append(alt)
    random.shuffle(stokket)
    
    qi = 1
    for alt in stokket:
        print(f"({qi}): {unescape(alt)}")
        qi += 1
    ans = int(input("Ditt svar:\n"))
    if stokket[ans-1] == korrekt:
        print("Det stemmer!")
        antall_riktige += 1
    else:
        print("Det er feil...")
        print("Korrekt svar er:", unescape(korrekt))

print(f"Du klarte {antall_riktige} av {n} spørsmål")
```
