```python
import numpy as np
import matplotlib.pyplot as plt

# Spiral parametere
theta = np.linspace(0, 8*np.pi, 1000)
r = np.linspace(0, 1, 1000)

x = r * np.cos(theta)
y = r * np.sin(theta)

plt.figure(figsize=(6,6))
plt.scatter(x, y, c=theta, cmap='plasma', s=10)
plt.axis("off")
plt.title("Spiral", fontsize=16)
plt.show()
```
