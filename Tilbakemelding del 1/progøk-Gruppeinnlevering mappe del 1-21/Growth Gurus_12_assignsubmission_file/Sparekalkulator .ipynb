```python
import math

# Definisjon av hvordan man kan beregne hvor mye Per må spare for å nå sitt sparemål
def beregn_maanedlig_sparing(sluttbelop, rente, antall_aar):
    n = 12  # Antall innskudd per år, dette betyr et innskudd i måneden
    r = rente / 100  # Rente omgjort til rente i desimaltall
    P = sluttbelop * (r / n) / ((1 + r / n) ** (n * antall_aar) - 1)
    return P 

# Definisjon av hvordan man kan beregne hvor lang tid det tar å nå et spare mål med et oppgitt beløp (Vi prøver med 10 000kr i måneden)
def beregn_sparetid(sluttbelop, maanedlig_sparing, rente):
    n = 12  # Antall innskudd per år, dette betyr innskudd i måneden
    r = rente / 100  # Rente omgjort til desimal
    t = math.log(1 + (sluttbelop * (r / n)) / maanedlig_sparing) / (n * math.log(1 + r / n))
    return t

# Funksjon for å justere månedlig sparing hvis renten endrer seg
def juster_for_renteendring(sluttbelop, rente_ny, antall_aar):
    return beregn_maanedlig_sparing(sluttbelop, rente_ny, antall_aar)

# Oppsummering av viktig informasjon
sluttbelop = 500_000  # Beløpet Per ønsker å spare sett bort ifra inflasjon
antall_aar = 3  # Antall år Per skal spare
rente = 5 * (1 - 0.37)  # 37% skatt på rentene
inflasjon = 0.02  # Inflasjon på 2% per år

# Man må justere for inflasjon
sluttbelop_justert = sluttbelop * (1 + inflasjon) ** antall_aar

# Løsning på den første delen av oppgaven (Hvor mye må Per spare i måneden for å nå sparemålet sitt i løpet av 3 år)
maanedlig_sparing = beregn_maanedlig_sparing(sluttbelop_justert, rente, antall_aar)
print(f"Per må spare {maanedlig_sparing:.2f} kr per måned for å nå {sluttbelop_justert:.2f} kr på {antall_aar} år.")

# Løsning på den andre delen av oppgaven (Hvor lang tid vil det ta Per å nå sparemålet sitt om han sparer 10 000kr i måneden)
maanedlig_belop = 10_000
sparetid = beregn_sparetid(sluttbelop_justert, maanedlig_belop, rente)
print(f"Hvis Per sparer 10 000 kr per måned, vil det ta Per {sparetid:.2f} år å nå målet.")

# Løsning på den tredje delen av oppgaven (Hvor mye må Per spare per måned om renten hadde vært satt til 6%)
rente_ny = 6 * (1 - 0.37)  # Rente etter skatt
maanedlig_sparing_ny = juster_for_renteendring(sluttbelop_justert, rente_ny, antall_aar)
print(f"Hvis renten øker til 6%, må Per spare {maanedlig_sparing_ny:.2f} kr per månedd for å nå målet.")
```
