```python
def beregn_populasjon(start_populasjon, vekstrate, år):
    # Formel for populasjonsberegningen
    return start_populasjon * ((1 + vekstrate) ** år)

def populasjonsprogram():
    # Viktig informasjon fra oppgaven
    område = "Ålesund"
    vekstrate = 0.02  # Dette betyr 2% hvert år
    startår = 2024
    sluttår = 2030
    start_populasjon = 58989

    # Beregner antall år "undersøkelsen" varer
    antall_år = sluttår - startår

    # Beregner populasjon ved sluttåret
    slutt_populasjon = beregn_populasjon(start_populasjon, vekstrate, antall_år)

    # Vekst i antall mennesker og i prosent
    vekst_i_antall_mennesker = slutt_populasjon - start_populasjon
    vekst_i_prosent = (vekst_i_antall_mennesker / start_populasjon) * 100

    # Svar på hvor mange mennesker som bor i Ålesund ved sluttåret
    print(f"Populasjon ved sluttår ({sluttår}): {int(slutt_populasjon)}")

    # Svar på vekst i antall folk og vekst i prosent
    print(f"Vekst i antall folk: {int(vekst_i_antall_mennesker)} personer")
    print(f"Vekst i prosent: {vekst_i_prosent:.2f}%")

# Vi må kjøre programmet
populasjonsprogram()
```
