# Oppgave 1.1 Populasjonsvekst
Dette programmet regner ut populasjonsveksten til et område ved hjelp av funksjoner og den logistiske ligningen. Hensikten er å se populasjons utviklingen i et område, viser vekst i antall, prosent og ved hjelp av grafisk fremstilling.

# Kode oppgave 1.1

```python
#Først må vi importere noen bibliotek som skal benyttes i koden.
import math 
import numpy as np

Land = input("område")
r = float(input("vekstrate")) #vekstrate. Må gjør eom til float for å kunne bruke videre i koden.
start_år = float(input("start_år"))
slutt_år = float(input("slutt_år"))
P0 = float(input("start_populasjon"))
K= float(input("maks_populasjon"))
t = int(slutt_år- start_år+1)
e = math.exp(1) 

#Definerer variabler programmet skal bruke i beregningene.
def regneutA(K,P0):
    A = (K-P0)/P0
    return A

#Definerer den funksjonen som skal foreta beregningen av populasjonsvekst.
def logistiske_ligningen(r,t,e,K,P0):
    A = regneutA(K,P0)
    Populasjons_vekst = K/(1+A*e**(-r*t))
    return Populasjons_vekst 
    
#Importerer flere bibliotek som gjør at programmet senere kan visualisere.
import matplotlib.pyplot as plt

#lag liste her? + plt.grafer

tid = list(range(t))
#+1 så vi tar med siste året, så alltdi med slutt_år

tid_populasjonsvekst=list(range(int(start_år), int(slutt_år)+1))

#Lager tom liste for populasjonsvekst. Dette skal senere legges til med append
populasjonsvekst = []
for t in tid:
    popv = logistiske_ligningen(r,t,e,K,P0)
    populasjonsvekst.append(popv)

Slutt_pop = logistiske_ligningen(r,t,e,K,P0)
print(Land, "har populasjonsveksten", Slutt_pop)

#Finner hvor mye populasjonsveksten har økt i antall mennesker og i prosent.
auke_folk = Slutt_pop- P0
print("auke_folk",auke_folk)

auke_prosent=(P0/Slutt_pop)*100
print("auke_prosent",auke_prosent)


# Visualiserer ved hjelp av matplotlib - bilbioteket. 
plt.plot(tid_populasjonsvekst, populasjonsvekst, label='populasjonsvekst', color = 'pink')

print(tid)
print(populasjonsvekst)

plt.xlabel('år')
plt.ylabel('populajson')
plt.title('populasjonsvekst')
plt.axhline(y=populasjonsvekst[-1], color='g', linestyle='--', label='Sluttpopulasjon')
plt.axhline(y = populasjonsvekst[0], color='r',linestyle='--', label='startpopulasjon')
plt.xticks(ticks = np.arange(start_år,slutt_år+1,1))
plt.legend()
plt.show()
```

# Diskusjon og resultat - oppgave 1.1
Programmet tar i bruk funskjonene for å printe ut populasjonen ved sluttår, hvor mye den har vokst i antall og hvor mye den har vokst i prosent. Den tar deretter funskjonen som regner ut sluttåret(logistiske ligningen) til å lage en liste og en for-løkke for å lage en graf som viser veksten per år.

# Oppgave 2.2 - Sparekalkulator
Denne koden ler en sparekalkulator. Programmet skal spørre bruker om noen verdier som senere skal bli benyttet i utregningen. Programmet vil beregne og oppgi sluttbeløpet på sparekontoen, og angi hvor mye av sluttbeløpet som er renter og innskudd. 

# Kode oppgave 2.2

```python
#Importerer bibliotek
import matplotlib.pyplot as plt

#Definerer variabler, bruker input-funksjonen.
P = float(input("Hva er terminbeløpet?: "))
R = float(input("Hva er renten i prosent: "))/ 100 #Gjør om rente til desimaltall
N = float(input("Hvor mange innskudd?: "))
T= float(input("Hvor mange år: "))

def saldo(P,R,N,T): 
    Slutt_beløp = P*((((1+R/N)**(N*T))-1)/(R/N)) 
    return Slutt_beløp

#Funksjon for totalinnskuddet 
def Total_innskuddlikning(P,N,T): 
    Total_innskudd = P*N*T #Viser summen av alle innskuddene 
    return Total_innskudd

#Funskjon som viser hvor mye av sluttbeløpet som er renter i kroner. 
def Rentekr_likning(Slutt_beløp, Total_innskudd): 
    Rente_i_kr = Slutt_beløp-Total_innskudd #Renteandel i kroner 
    return Rente_i_kr

#Funskjon som viser renten i % av sluttbeløpet: 
def Renteprosent(Rente_i_kr, Slutt_beløp): 
    Rente_i_pro = (Rente_i_kr/Slutt_beløp)*100 #Renten i %
    return Rente_i_pro

#Funsksjon som viser innskuddets prosentandel av sluttbeløpet: 
def Innskuddsprosent(Total_innskudd, Slutt_beløp): 
    Innskuddsprosent = (Total_innskudd/ Slutt_beløp) * 100 
    return Innskuddsprosent

#Beregner sluttbeløpet 
Slutt_beløp = saldo(P,R,N,T) 
print("Sluttbeløpet på sparekontoen er:", round(Slutt_beløp, 2))

#Beregner det totale innskudd 
Total_innskudd = Total_innskuddlikning(P,N,T) 
print("Det totale innskuddet er: ",Total_innskudd)

#Beregner rentekroner 
Rente_i_kr = Rentekr_likning(Slutt_beløp, Total_innskudd) 
print("Renten i kroner beregnet av sluttbeløp er: ",round(Rente_i_kr, 2))

#Beregner renten i %
Rente_i_pro = Renteprosent(Rente_i_kr, Slutt_beløp) 
print("Renten i prosent beregnet av sluttbeløpet er: ", round(Rente_i_pro, 2), "%")

#Beregner innskuddsprosenten
Innskuddsprosent = Innskuddsprosent(Total_innskudd, Slutt_beløp) 
print("Innskuddenes prosentandel beregnet av sluttbeløpet er: ", round(Innskuddsprosent, 2), "%")

#Bruker en if-else setning for å regne ut for brukeren om hen har råd til det hen vil spare opp til. 
spare_til = input("Hva ønsker du å spare til? ")
Beløpet = int(input("Hvor mye koster det? "))
if Slutt_beløp > Beløpet: print("Supert! Du har spart opp til",spare_til,"med",round(Slutt_beløp-Beløpet),"kroner til overs") 
else: 
    print("Dessverre, du mangler ",int(Beløpet-Slutt_beløp),"kroner")
    
#Simulerer ved hjelp av et stolpediagram
labels = ['Sluttbeløp', 'Kostnad']
values = [Slutt_beløp, Beløpet] 

plt.bar(labels, values, color=['orange', 'green'])  #Farger til stolpediagrammet
plt.title('Sammenligning mellom sluttbeløp og kostnad for ønsket')  # Navnet på diagrammet
plt.ylabel('Beløp i kroner')  # Y-akse tittel

#Legger til verdier
for i, value in enumerate(values):
    plt.text(i, value + 100, f'{round(value, 2)} NOK', ha='center', va='bottom')

plt.show()  #Simulerer programmet
```

# Resultat og diskusjon oppgave 2.2
Programmet spør brukeren om terminbeløp, renten, hvor mange innskudd og hvor mange år. den tar i bruk variablene for å sette den inn i Funskjonene for å regne ut sluttbeløpet, total innskudd, renten i kroner, renten i prosent og innskuddens prosetnandel av sluttbeløpet.
Til slutt spør den brukeren hva hen vil spare til for å regne ut om brukeren har råd til det. Dette gjør den ved å spør brukeren hvor mye det koster og ved bruk av if-else funskjonen. Har brukeren råd så vil den returnere at brukeren har råd og hvor mye den har til overs. Hvis hen ikke har råd så vil den returnere at hen ikke har råd og hvor mye hen mangler.

# Oppgave 5.1 - Simuleringsoppgave
Denne koden skal ta for seg hvordan omsetningen endres når det er salg på noen varer. Programmet skal se hvor stor endring dette blir ut ifra hvor stor kjøpssasnnsynligheten er knyttet til hvert produkt når det går på salg, sammenlignet med normalpris. For å få det til å fungere må vi først foreta noen antagelser. Programmet tar utgangspunkt i en måned, og ser hvordan å sette varene på salg påvirker omsetning og salg.

# Kode 5.1

```python
#Først og fremst må vi importere noen biblioteker
import matplotlib.pyplot as plt
import random
import numpy as np

#Finner opp noen varer og hva prisen deres er i butikk
produkt_vanligpris = [
    {"navn" : "Shorts", "pris": 500, "kjopsannsynlighet": 0.4},
    {"navn" : "Hettegenser", "pris": 400, "kjopsannsynlighet": 0.2},
    {"navn" : "T-skjorte", "pris": 250, "kjopsannsynlighet": 0.3},
    {"navn" : "Utejakke", "pris": 700, "kjopsannsynlighet": 0.1},
] 

#Varer på salg øker kjøpssannsynlighetn med 20 prosent
produkter_pa_salg = [
    {"navn" : "Shorts", "pris": 400, "kjopsannsynlighet": 0.6},
    {"navn" : "Hettegenser", "pris": 300, "kjopsannsynlighet": 0.4},
    {"navn" : "T-skjorte", "pris": 150, "kjopsannsynlighet": 0.5},
    {"navn" : "Utejakke", "pris": 600, "kjopsannsynlighet": 0.3},
]

#Antakelse over antall kunder innom butikken per dag
antallkunder_en_dag = 50
dager_antall = 30 #i én måned


#Funksjon som simulere ett kunde-besøk
def besok_av_kunde(produkter):
    budsjett = random.uniform(500, 2000) #Fine et tilfeldig budsjett for kunden mellom 500kr - 2000kr
    sammenlagt_omsetning = 0
    antall_solgt = 0

#Kunden ser gjennom de ulike produktene
    for produkt in produkter:
        pris = produkt["pris"]
        sannsynlighet = produkt["kjopsannsynlighet"]
        
        #Sjekker om kunden kjøper produktet, og om kunden har budsjett til kjøpet
        if random.random() < sannsynlighet:
            if budsjett >= pris:
                sammenlagt_omsetning += pris  #Øker omsetningen med den prisen som kunden betaler.
                antall_solgt += 1 #Hver gang en kunde kjøper et produkt, øker antall_salg med 1. Dette sporer hvor mange produkt som blir kjøpt. 
    
    return sammenlagt_omsetning, antall_solgt


#Funksjon som simulerer kjøp for én måned
def simulerer_maned(produkter):
    sammenlagt_omsetning_mnd = 0  #Setter verdiene = 0 fordi ved starten er det 0, men dette kommer til å endres på. 
    total_salg_mnd = 0
  #Simulerer hver dag
    for dag in range(dager_antall):   #Range forteller hvor mange ganger forløkken skal gå.
        omsetning_dag = 0
        antall_salg_dag = 0

        #Simulerer for alle kundene som besøker butikken
        for kunde in range(antallkunder_en_dag):
            omsetning, antall_solgt = besok_av_kunde(produkter)
            omsetning_dag += omsetning #summerer omsetningen fra dag til dag
            antall_salg_dag += antall_solgt #summerer antall salg fra dag til dag

        #Summerer omsetning og salg for dagen til månedssummen
        sammenlagt_omsetning_mnd += omsetning_dag
        total_salg_mnd += antall_salg_dag

    return sammenlagt_omsetning_mnd, total_salg_mnd

#Simulerer én måned uten salg
omsetning_normal, salg_normal = simulerer_maned(produkt_vanligpris)

#Simulerer én måned med salg
omsetning_salg, salg_salg = simulerer_maned(produkter_pa_salg)

#Resultater
print(f"Omsetningen uten salg: {omsetning_normal} kr, Antall solgte varer er: {salg_normal}")
print(f"Omsetningen med salg: {omsetning_salg} kr, Antall solgte varer er: {salg_salg}")

#Fremstiller resultatet visuelt i et stolpediagram
plt.bar(["Uten salg", "Med salg"], [omsetning_normal, omsetning_salg], color=['lightgreen', 'orange'])
# Legger til omsetningen i kr på stolpene
plt.text(0, omsetning_normal, f"{omsetning_normal:.2f} kr", ha='center', va='bottom')
plt.text(1, omsetning_salg, f"{omsetning_salg:.2f} kr", ha='center', va='bottom')
plt.title("Omsetningen med og uten salg")
plt.show()
```

# Resultat og diskusjon oppgave 5.1
Koden viser at omsetningen øker betydelig når varer er på salg, sammenlignet med perioder uten salg. Dette samsvarer med forventningen om at rabatter øker kunders kjøpslyst. Programmet visualiserer også forskjellen i omsetning for en måned med og uten salg ved hjelp av et stolpediagram.
