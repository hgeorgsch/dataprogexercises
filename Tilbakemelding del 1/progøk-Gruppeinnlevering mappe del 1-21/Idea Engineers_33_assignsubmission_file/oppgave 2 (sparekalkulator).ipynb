```python
* Lag et program som viser og plotter ulike sammenhenger mellom parametre under sparing:
    - Sluttsaldo etter x (feks x=10) år som funksjon av rente
    - Sluttsaldo etter x (feks x=10) år med fast terminbeløp $P$ (feks = $P=1000$) som funksjon av antall innskudd i året
    - Andel renter som funksjon av tid, og andel renter etter x år som funksjon av rentesats
```

```python
#Oppg 3.2.2
import matplotlib.pyplot as plt

# Bruker input
terminbeløp = float(input("Størrelse på terminbeløp kr: "))
spare_tid = int(input("Hvor mange år vil du spare?: "))
rente = float(input("Legg inn årlig rentesats i desimalform: ")) 
n = int(input("Antall innskudd per år: "))  

# Spareformelen
def beregn_sparing(P, r, n, t):
    fremtidig_verdi = P * ((1 + r/n) ** (n*t) - 1) / (r/n)
    return round(fremtidig_verdi, 2)

# plotting av sluttsaldo som funksjon av rente
def plot_saldo_vs_rente(P, n, t):
    rentesatser = [i / 100 for i in range(1, 21)] #renteliste
    saldoer = [beregn_sparing(P, r, n, t) for r in rentesatser] #liste med sparing etter t antal år

    plt.plot([r*100 for r in rentesatser], saldoer, marker='o', linestyle='--')
    plt.xlabel("Rentesats [%]")
    plt.ylabel(f"Sluttsaldo etter {t} år [kr]")
    plt.title(f"Sluttsaldo etter {t} år som funksjon av rente med P={P} kr")
    plt.grid()
    plt.show()


# plotter sluttsaldo som funksjon av antall innskudd i året
def plot_saldo_vs_innskudd_per_år(P, r, t):
    innskudd_antall = list(range(1, 13)) 
    saldoer = [beregn_sparing(P, r, n, t) for n in innskudd_antall]

    plt.plot(innskudd_antall, saldoer, marker='x', linestyle='-')
    plt.xlabel("Antall innskudd per år")
    plt.ylabel(f"Sluttsaldo etter {t} år [kr]")
    plt.title(f"Sluttsaldo med fast terminbeløp som funksjon av antall innskudd i året")
    plt.grid()
    plt.show()

# plotter andel renter som funksjon av tid
def plot_andel_renter(P, r, n, t):
    saldo = 0
    total_renter = 0
    år_liste = list(range(1, t + 1))
    renter_andel_liste = []

    for år in år_liste:
        renter = saldo * r / n
        total_renter += renter
        saldo += P + renter
        renter_andel = total_renter / saldo if saldo > 0 else 0
        renter_andel_liste.append(renter_andel)

    plt.plot(år_liste, renter_andel_liste, marker='s', linestyle='-.', color='green')
    plt.xlabel("År")
    plt.ylabel("Andel renter")
    plt.title(f"Andel renter som funksjon av tid")
    plt.grid()
    plt.show()


# funksjonskall for å beregne fremtidig saldo
fremtidig_saldo = beregn_sparing(terminbeløp, rente, n, spare_tid)
print(f"Fremtidig saldo etter {spare_tid} år: {fremtidig_saldo} kr")

# Plotter
plot_saldo_vs_rente(terminbeløp, n, spare_tid)
plot_saldo_vs_innskudd_per_år(terminbeløp, rente, spare_tid)
plot_andel_renter(terminbeløp, rente, n, spare_tid)


""" har laga et program med:
    - funksjon som regner ut sparing
    -3 forskjellige funksjoner som plotter inn renteandel, saldo etter antall inskudd og sluttsaldo 
    - brukte lister for å kompilere grafene
    - brukte f-streng for å få ting på rett plass
    - Funksjonskall for å kjøre alt
```
