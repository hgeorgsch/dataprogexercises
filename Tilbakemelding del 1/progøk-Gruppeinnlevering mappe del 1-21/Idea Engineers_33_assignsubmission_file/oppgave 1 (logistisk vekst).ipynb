## Oppgave 3 - populasjonsvekst 
* Denne oppgaven tar for seg Haram kommune og Sula kommune, og sammenligner befolkningsveksten i de to kommunene fra 2020 til 2040

```python
import math
import matplotlib.pyplot as plt

# Funksjon for å beregne populasjon ved tid t
def populasjonsvekst(K, A, r, t):
    return K / (1 + A * math.exp(-r * t))

def simuler_vekst(navn, vekstrate, startår, sluttår, start_populasjon, makspopulasjon):
    t = sluttår - startår
    A = (makspopulasjon - start_populasjon) / start_populasjon
    
    år = list(range(startår, sluttår + 1))
    populasjoner = [populasjonsvekst(makspopulasjon, A, vekstrate, t - startår) for t in år]
    
    slutt_populasjon = populasjoner[-1]
    absolutt_vekst = slutt_populasjon - start_populasjon
    relativ_vekst = (absolutt_vekst / start_populasjon) * 100
    
    info = {
        "navn": navn,
        "start_populasjon": start_populasjon,
        "slutt_populasjon": slutt_populasjon,
        "absolutt_vekst": absolutt_vekst,
        "relativ_vekst": relativ_vekst,
        "populasjoner": populasjoner,
        "år": år
    }
    
    return info

def sammenligne_områder(områder):
    # Plotting av utvikling for alle områder
    plt.figure(figsize=(10, 6))
    
    for område in områder:
        plt.plot(område["år"], område["populasjoner"], marker='o', linestyle='-', label=område["navn"])
    
    plt.title("Populasjonsutvikling for flere områder")
    plt.xlabel("År")
    plt.ylabel("Populasjon")
    plt.legend()
    plt.grid(True)
    plt.show()
    
    # Sammenligne vekst og utskrift av resultater
    størst = max(områder, key=lambda x: x["slutt_populasjon"])
    mest_vekst = max(områder, key=lambda x: x["absolutt_vekst"])
    
    print("\nSammenligning av befolkningsvekst:")
    for område in områder:
        print(f"\nOppsummering for {område['navn']}:")
        print(f"Populasjonen ved startåret: {område['start_populasjon']:.2f}")
        print(f"Populasjonen ved sluttåret: {område['slutt_populasjon']:.2f}")
        print(f"Absolutt vekst: {område['absolutt_vekst']:.2f} personer")
        print(f"Relativ vekst: {område['relativ_vekst']:.2f}%")
    
    print(f"\nOmrådet med størst populasjon i sluttåret er {størst['navn']} med {størst['slutt_populasjon']:.2f} personer.")
    print(f"Området med mest absolutt vekst er {mest_vekst['navn']} med en vekst på {mest_vekst['absolutt_vekst']:.2f} personer.")

def main():
    n = int(input("Hvor mange områder vil du simulere?: "))
    områder = []
    
    for i in range(n):
        print(f"\nOmråde {i + 1}:")
        navn = input("Hva heter området du vil undersøke? ")
        vekstrate = float(input("Hva er vekstraten (i prosent): ")) / 100
        startår = int(input("Når er startåret: "))
        sluttår = int(input("Når er sluttåret: "))
        start_populasjon = float(input("Populasjon ved startår: "))
        makspopulasjon = float(input("Hva er makspopulasjon i området: "))
        
        info = simuler_vekst(navn, vekstrate, startår, sluttår, start_populasjon, makspopulasjon)
        områder.append(info)
    
    sammenligne_områder(områder)

if __name__ == "__main__":
    main()
```
