# Populasjonsvekst

 * Programmet spør bruker etter:
   - Navn på området (eg. Island, Finmark osv)
    - Vekstrate
    - Startår
    - Sluttår
    - Populasjon ved startår
    - Makspopulasjon
* Programmet beregner populasjonsstørrelse ved sluttår
* Programmet oppsummerer og gir tilbakemelding om størrelsene som er gitt
* Programmet angir populasjon ved sluttår
* Programmet angir hvor mye populasjonen har vokst absolutt (antall folk) og relativt (prosent)

Den logistiske ligningen er gitt ved:
$$
P(t) = \frac{K}{1+Ae^{-rt}}
$$
Her er:
* $P(t)$: Populasjon ved tiden t
* $t$ er tiden
* $r$ er relativ vekstrate, feks 2%
* $K$ er makskapasiteten til populasjonen
* $e$ er eulers tall: $e\approx 2,71828$
  
$A$ er en koeffisient definert ved:
$$
A = \frac{K-P_0}{P_0}
$$
Hvor $P_0$ er populasjonen ved $t=0$

```python
import math
import matplotlib.pyplot as plt

navn_paa_omraadet = input("Navn på området: ")
vekstrate = float(input("Vekstraten i dette området er (i prosent, f.eks. 2 for 2%): ")) / 100 
startaar = int(input("Startår: "))
sluttaar = int(input("Sluttår: "))
populasjon_startaar = int(input("Populasjonen ved startår: "))
makskapasitet_populasjon = int(input("Makskapasiteten til populasjonen er: "))

t = (sluttaar - startaar)
r = vekstrate
P0 = populasjon_startaar
K = makskapasitet_populasjon
A = (K - P0) / P0

Populasjon_sluttaar = K / (1 + A * math.exp(-r * t))
Populasjon_sluttaar_avrundet = round(Populasjon_sluttaar)

prosentvis_vekst = ((Populasjon_sluttaar - populasjon_startaar) / populasjon_startaar) * 100
prosentvis_vekst_avrundet = int(round(prosentvis_vekst))

print("\nSluttpopulasjon:", Populasjon_sluttaar)
print("Sluttpopulasjon avrundet:", int(Populasjon_sluttaar_avrundet))
print("\nNår du oppgir opplysningene om populasjonen på", navn_paa_omraadet, "med", startaar,
      "som startår, og med en vekstrate på", vekstrate * 100, "%, \nvil populasjonen ved sluttåret", sluttaar,
      "være på ca:", int(Populasjon_sluttaar_avrundet), "som gir en prosentvis økning på ca:", prosentvis_vekst_avrundet, "%.")

aar = list(range(startaar, sluttaar + 1)) 
populasjoner = [] 

for t in aar:
    P_t = K / (1 + A * math.exp(-r * (t - startaar)))
    populasjoner.append(P_t)

plt.figure(figsize=(10, 6))
plt.plot(aar, populasjoner, label='Populasjonsvekst', color='green')
plt.title(f'Populasjonsvekst i {navn_paa_omraadet}')
plt.xlabel('År')
plt.ylabel('Populasjon')
plt.axhline(y=makskapasitet_populasjon, color='red', linestyle='--', label='Makspopulasjon')
plt.legend()
plt.grid()
plt.xticks(aar)
plt.tight_layout()
plt.show()
```

# Resultat og diskusjon
Jeg har opprettet et program som spør bruker om opplysninger angående populasjonsvekst. Ved å ta opplysningene fra min hjemkommune, Vanylven, ser vi at populasjonsveksten er liten, og en positiv vekst i befolkningen øker sakte. Her kan bruker velge selv hvilke områder den vil sjekke, og programmet vil oppsummere og vise resultatet grafisk. 

# Sparekalkulator

Dersom vi sparer et fast terminbeløp $P$ som vi betaler inn på konto $n$ ganger i året, vi har en p.a. rente på $r$ prosent og vi sparer i $t$ år, vil sluttbeløpet på sparekontoen være gitt ved:
$$
F = P\cdot \left(\frac{\left(1+\frac{r}{n}\right)^{nt}-1}{\frac{r}{n}}\right)
$$

Lag et program som ber bruker av programmet om å taste inn terminbeløp, rente, antall innskudd/renteavregninger og antall år og gir følgende tilbakemeldinger:
* Programmet oppsummerer og gir tilbakemelding på tallene brukeren gav
* Programmet beregner og oppgir sluttbeløpet på sparekonto
* Programmet angir hvor mye av sluttbeløpet er renter og innskudd, i prosent og kroner

```python
import math
import matplotlib.pyplot as plt

terminbelop = float(input("Terminbeløp (kr): "))
rente = float(input("Rente (i prosent): ")) / 100 
innskudd = int(input("Antall innskudd/renteavregninger i året: "))
antall_aar = int(input("Antall år: "))

P = terminbelop
n = innskudd
r = rente
t = antall_aar

Slutt_belop = P * ((((1 + r/n)**(n*t)) - 1) / (r/n))
Slutt_belop_avrundet = round(Slutt_belop)

print("\nMed", innskudd, 
      "innskudd i året på", terminbelop, "kr hver, og en rentesats på", rente*100, "%, blir sluttbeløpet på sparekonto etter", t,
      "år, omtrent", Slutt_belop_avrundet, "kr.\n")

formueskatt = Slutt_belop * 0.01
formueskatt_avrundet = round(formueskatt, 2)
print("Formueskatt:", formueskatt_avrundet, "kr")

renteinntekt = Slutt_belop - (P * n * t)
renteinntekt_avrundet = round(renteinntekt, 2)
print("Opptjente renteinntekter:", renteinntekt_avrundet, "kr")

skatteprosent = renteinntekt * 0.22
skatteprosent_avrundet = round(skatteprosent, 2)
print("Skatt:", skatteprosent_avrundet, "kr (22% av renteinntekter)")

prosentoke_konto = (1 - (P * n * t) / Slutt_belop) * 100
prosentoke_avrundet = round(prosentoke_konto, 2)
print("\nMed et innskudd på til sammen", P * n * t, "kr, og et sluttbeløp på", Slutt_belop_avrundet, "kr, \nvil dette si at beløpet på konto har økt med omlag", 
      prosentoke_avrundet, "% i renter i løpet av", t, "år, og gir en renteinntekt på omtrent", int(round(renteinntekt_avrundet)), "kr.")

aar = list(range(1, t + 1))
innskudd_over_tid = []
renter_over_tid = []

for aar_i in aar:
    F = P * ((((1 + r/n)**(n * aar_i)) - 1) / (r/n))
    innskudd_over_tid.append(P * n * aar_i)
    rente_for_aar = F - (P * n * aar_i)
    renter_over_tid.append(rente_for_aar)

plt.figure(figsize=(10, 6))
plt.bar(aar, innskudd_over_tid, label='Totalt innskudd', color='green')
plt.bar(aar, renter_over_tid, bottom=innskudd_over_tid, label='Renter opptjent', color='blue')
plt.title('Utvikling av innskudd og renter over tid')
plt.xlabel('År')
plt.ylabel('Beløp (kr)')
plt.xticks(aar)
plt.legend()
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

# Resultat og diskusjon
I denne deloppgaven kan bruker selv føre inn verdier ut fra ønsket resultat i sparekalkulator. Her føres inn ønsket terminbeløp, antall innskudd i året, og antall ønskede år, samt kontoens renter i prosent for å regne ut renteinntektene. Programmet oppsummerer resulatet, og viser grafisk summen oppspart som inkluderer både innskuddet og renteinntektene hvert år. 

# Kundeatferd

 Vi ønsker å undersøke hvordan det å sette varer på tilbud påvirker salg og omsetning

Anta at en butikk ar et knippe produkter til en viss pris

Når en kunde kommer inn i butikken er det en viss sjanse knyttet til hvert produkt for at kunden kjøper det

Dersom varene kommer på salg øker denne sjansen

Gjør noen antagelser om hvor mange kunder som kommer innom butikken hver dag

Når en kunde kommer inn i butikken har de et visst budjsett som vi trekker tilfeldig fra en passende fordeling

Her kan dere bruke feks random.gauss(),numpy.random.normal(), random.uniform()
Kunden ser så gjennom produktene og kjøper de basert på tilfeldighet (produktets parametre) og om de har penger igjen i budsjettet

Sammenlign salg og omsetning for en måned med og uten produkter på tilbud

Du velger selv hvordan sjansen for salg skal øke med tilbudsrate -- men den logistiske modellen er en god kandidat

Fremstill resultatet av simulering grafisk med matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt

antall_dager = 30
antall_kunder_per_dag = 100
antall_produkter = 5
priser = np.array([20, 30, 40, 50, 60])
basis_sjanser = np.array([0.1, 0.15, 0.2, 0.25, 0.3])
rabatt_rate = 0.5
rabatterte_sjanser = basis_sjanser * (1 + rabatt_rate)
budsjett_gjennomsnitt = 50
budsjett_standardavvik = 15

omsetning_uten_rabatt = []
omsetning_med_rabatt = []

for dag in range(antall_dager):
    daglig_omsetning = 0
    for kunde in range(antall_kunder_per_dag):
        budsjett = np.random.normal(budsjett_gjennomsnitt, budsjett_standardavvik)
        for pris, basis_sjanse in zip(priser, basis_sjanser):
            if np.random.rand() < basis_sjanse and budsjett >= pris:
                daglig_omsetning += pris
                budsjett -= pris
    omsetning_uten_rabatt.append(daglig_omsetning)

for dag in range(antall_dager):
    daglig_omsetning = 0 
    for kunde in range(antall_kunder_per_dag):
        budsjett = np.random.normal(budsjett_gjennomsnitt, budsjett_standardavvik)
        for pris, rabattert_sjanse in zip(priser, rabatterte_sjanser):
            if np.random.rand() < rabattert_sjanse and budsjett >= pris:
                daglig_omsetning += pris
                budsjett -= pris
    omsetning_med_rabatt.append(daglig_omsetning) 

plt.figure(figsize=(12, 6))
plt.plot(omsetning_uten_rabatt, label='Salg uten tilbud', color='blue')
plt.plot(omsetning_med_rabatt, label='Salg med tilbud', color='green')
plt.title('Salg og omsetning: Med vs. Uten tilbud')
plt.xlabel('Dager')
plt.ylabel('Omsetning')
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()  
```

# Resultat og diskusjon
Programmet viser grafisk kundeatferd ved salg av produkt med og uten tilbud. I grafen ser vi svingninger i omsetning, med flest omsetning av salg med tilbud, foruten om ved ca. dag 3 og dag 12, da salg uten tilbud var høyere omsetning av.
