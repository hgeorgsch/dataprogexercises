Populasjonsveksten antas beskrevet av den logistiske ligningen som sist:

$$
P(t) = \frac{K}{1+Ae^{-rt}}
$$
Her er:
* $P(t)$: Populasjon ved tiden t
* $t$ er tiden
* $r$ er relativ vekstrate, feks 2%
* $K$ er makskapasiteten til populasjonen
* $e$ er eulers tall: $e\approx 2,71828$

$A$ er en koeffisient definert ved:
$$
A = \frac{K-P_0}{P_0}
$$
Hvor $P_0$ er populasjonen ved $t=0$

# Oppgave 2.2: Sparekalkulator
----

Dersom vi sparer et fast terminbeløp $P$ som vi betaler inn på konto $n$ ganger i året, vi har en p.a. rente på $r$ prosent og vi sparer i $t$ år, vil sluttbeløpet på sparekontoen være gitt ved:
$$
F = P\cdot \left(\frac{\left(1+\frac{r}{n}\right)^{nt}-1}{\frac{r}{n}}\right)
$$

Lag et program som ber bruker av programmet om å taste inn terminbeløp, rente, antall innskudd/renteavregninger og antall år og gir følgende tilbakemeldinger:
* Programmet oppsummerer og gir tilbakemelding på tallene brukeren gav
* Programmet beregner og oppgir sluttbeløpet på sparekonto
* Programmet angir hvor mye av sluttbeløpet er renter og innskudd, i prosent og kroner

*(Merk at $r$ i formelen ikke er i prosent, 2% = 0.02)*

### Dersom du vil
----
Dersom du vil kan du også ta inflasjon, formueskatt ($\approx 1\%$) og skatteprosent (22% av renteinntekter) med i beregningene

```python

# Sparekalkulator
import math
import matplotlib.pyplot as plt
e=math.exp (1)

def beregn_sluttbelop(P, r, n, t):

   F = P*(  ( ((1+r/n)**(n*t))-1)   /  (r/n)  )
   return F


def juster_for_inflasjon(belop, inflasjon, aar):
    inflasjonsfaktor = (1 + inflasjon / 100) ** aar
    justert_belop = belop / inflasjonsfaktor
    return justert_belop

def beregn_formueskatt(belop, formueskatt):
    skatt = belop * formueskatt / 100
    return skatt
    

#Henter input fra bruker
P = float(input("Hva er terminbeløpet?"))
r = float(input("Hva er renten?")) / 100
n = int(input("Hvor mange ganger i året skal du sette inn penger?"))
t = int(input("Hvor mange år skal du spare?"))
inflasjon = float(input("Hva er årlige inflasjonsraten"))
formueskatt = float(input("Hva er satsen på formueskatte"))


#Beregn sluttbeløp
sluttbelop = beregn_sluttbelop(P, r, n, t)


#Regner ut sluttbeløp i prosent og kroner
resultat = round(P*((1+r/n)**(n*t)-1)/(r/n))
rente_prosent = round((resultat-P*n*t)/(P*n*t)*100)
innskudd_prosent = (100-rente_prosent)
innskudd_sum = P*n*t
rente_sum = (resultat-innskudd_sum)
andel_rente = (rente_sum/resultat) #Gå over å kall resultat for slutt_saldo


# Endre for inflasjon
justert_sluttbelop = juster_for_inflasjon(sluttbelop, inflasjon, t)
justerte_renter = juster_for_inflasjon(rente_sum, inflasjon, t)


# Beregn formueskatt
skatt = beregn_formueskatt(sluttbelop, formueskatt)
sluttbelop_etter_skatt = sluttbelop - skatt


#Utskrift av resultater
print("Hvis du setter in", P, "kr", n, "ganger i året, blir du å sitt igjen med", resultat, "kr", "etter", t, "år.")
print(f"andel renter er {andel_rente:.1%}")
print(f"Opptjente renter: {rente_sum:.2f} NOK")
print(f"Sluttbeløp justert for inflasjon: {justert_sluttbelop:.2f}NOK")
print(f"Opptjente renter justert for inflasjon: {justerte_renter:.2f} NOK")
print(f"Formueskatt: {skatt:.2f} NOK")
print(f"Sluttbeløp etter formueskatt: {sluttbelop_etter_skatt:.2f} NOK")


#Plotter en graf
aar_liste = list(range(1, t + 1))
saldo_liste = []
justert_saldo_liste = []
saldo_etter_skatt_liste = []

for i in aar_liste:
    saldo = beregn_sluttbelop(P, r, n, i)
    saldo_liste.append(saldo)
    justert_saldo = juster_for_inflasjon(saldo, inflasjon, i)
    justert_saldo_liste.append(justert_saldo)
    saldo_etter_skatt = saldo - beregn_formueskatt(saldo, formueskatt)
    saldo_etter_skatt_liste.append(saldo_etter_skatt)

plt.figure(figsize=(14, 10))
plt.subplot(2,2,1)
plt.plot(aar_liste, saldo_liste, label='Saldo uten justeringer')
plt.grid("True")
plt.xlabel('År')
plt.ylabel('Saldo (NOK)')
plt.title('Utvikling av sparebeløp over tid')

plt.figure(figsize=(14, 10))
plt.subplot(2,2,2)
plt.plot(aar_liste, justert_saldo_liste, label='Saldo justert for inflasjon')
plt.grid("True")
plt.xlabel('År')
plt.ylabel('Saldo justert for inflasjon (NOK)')
plt.title('Utvikling av sparebeløp over tid')

plt.figure(figsize=(14, 10))
plt.subplot(2,2,3)
plt.plot(aar_liste, saldo_etter_skatt_liste, label='Saldo etter formueskatt')
plt.grid("True")
plt.xlabel('År')
plt.ylabel('Saldo etter formueskatt (NOK)')
plt.title('Utvikling av sparebeløp over tid')

plt.tight_layout()

plt.legend()
plt.show()
```
