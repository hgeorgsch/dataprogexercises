```python
#Sparekalkulator
Her skal jeg lage et program som tar inn et sett med verdier fra brukeren som omhandler sparing, og skal deretter beregne og presentere sluttverdi på sparekonto, total innskudd og renter
```

```python
#lager en funsjon som kalkulerer sluttbeløp på sparekonto og returnerer det
def kalkuler_sluttbeløp(P, r, n, t):
    sluttbeløp = P * ((1 + r / n) ** (n * t) - 1) / (r / n)
    return sluttbeløp

#lager en funksjon som tar input fra brukeren, gjør relevante beregninger, og oppsumerer tallene gitt og resultatet av det
def sparekalkulator(): 
    P = float(input("terminbeløp i kr: "))
    r = float(input("rente p. a oppgitt i desimaltall: "))
    n = int(input("mengden inskudd pr. år: "))
    t = int(input("antall år med sparing: "))

    #kalkulering av sluttbeløp
    sluttbeløp = kalkuler_sluttbeløp(P, r, n, t)
    innskudd_totalt = P * n * t
    renter = sluttbeløp - innskudd_totalt

    #oppsumerer tall som ble gitt
    print(f"\nOppsummering av tall fra input:")
    print(f"Terminbeløp: {P} kr")
    print(f"Rente p.a.: {r*100:.2f}%")
    print(f"Mengden innskudd pr. år: {n}")
    print(f"Mengden år: {t}")
    
    # Printer ut alle resultatene av beregningene med de tallene som ble gitt
    print(f"\nSluttbeløp på sparekonto: {sluttbeløp:.2f} kr")
    print(f"Totalt innskudd: {innskudd_totalt:.2f} kr")
    print(f"Totale renter: {renter:.2f} kr")
    print(f"Renter i prosent av sluttbeløp: {(renter / sluttbeløp) * 100:.2f}%")

#til slutt kaller vi på sparekalkulator funksjonen for å kjøre programmet
sparekalkulator()

    
    
```
