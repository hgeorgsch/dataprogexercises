```python
#Populasjonsvekst
Her skal jeg lage et program som skal ta inn relevante tall for populasjonsvekst av et land og gjøre kalkuleringer basert på det, og deretter presentere dataene på en fin og oversiktelig måte.
```

```python
import math

def main():
    #her putter bruker inn relevant informasjon
    område = input("Hva er navnet på området?: ")
    vekstrate = float(input("Hva er vekstraten i prosent pr. år?: "))
    startår = int(input("Hva er startåret?: "))
    sluttår = int(input("Hva er sluttåret?: "))
    populasjon_start = int(input("Hva er populasjonen ved startår?: "))
    populasjon_max = int(input("Hva er makspopulasjonen?: "))

    #antall år
    t = sluttår-startår

    #beregn A
    A = (populasjon_max-populasjon_start)/populasjon_start

    #beregning av populasjonsstørrelse ved sluttår
    populasjon_slutt = populasjon_max / (1 + A*math.exp(-vekstrate*t))
    

    #beregning av absolutt vekst
    absolutt_vekst = populasjon_slutt-populasjon_start

    #beregning av relativ vekst
    relativ_vekst = (absolutt_vekst/populasjon_start)*100

    #oppsummering og resultater
    print("\n--- Populasjonsberegning for området:", område, "---")
    print(f"Startår: {startår}")
    print(f"Sluttår: {sluttår}")
    print(f"Populasjon ved startåret: {populasjon_start}")
    print(f"Maksimal populasjon: {populasjon_max}")
    print(f"Vekstrate: {vekstrate * 100}% per år")
    print(f"Populasjon ved sluttåret: {int(populasjon_slutt)}")
    print(f"Absolutt vekst i populasjonen: {int(absolutt_vekst)} personer")
    print(f"Relativ vekst i populasjonen: {relativ_vekst:.2f}%")
   
main()
```

```python
#Resultat
Koden tar inn tall som bruker putter inn for hver gang den kjøres, og presenterer alle tallene og beregningene programmet har gjort på slutten.
```
