```python
import matplotlib.pyplot as plt


def beregn_sluttbelop(terminbelop, rente, antall_innskudd, antall_ar):
    saldo_over_tid = []  # Liste for å lagre saldoen over tid
    sluttbelop = 0
    for i in range(antall_ar * antall_innskudd):
        sluttbelop = (sluttbelop + terminbelop) * (1 + rente / antall_innskudd)
        saldo_over_tid.append(sluttbelop)  # Legg til saldoen for hver termin
    return sluttbelop, saldo_over_tid


def oppsummering(terminbelop, rente, antall_innskudd, antall_ar, sluttbelop, totalt_innskudd, renteinntekter, renteandel):
    print("\nOppsummering:")
    print(f"Terminbeløp: {terminbelop} kr")
    print(f"Rente: {rente * 100}%")
    print(f"Antall innskudd per år: {antall_innskudd}")
    print(f"Antall år: {antall_ar}")
    print(f"Totalt spart beløp uten renter: {totalt_innskudd} kr")
    print(f"Sluttbeløp på sparekonto: {sluttbelop:.2f} kr")
    print(f"Renteinntekter: {renteinntekter:.2f} kr")
    print(f"Andel av sluttbeløpet som er renter: {renteandel:.2f}%")


def beregn_innskudd(terminbelop, antall_innskudd, antall_ar):
    return terminbelop * antall_innskudd * antall_ar


def beregn_renteandel(totalt_innskudd, sluttbelop):
    renteinntekter = sluttbelop - totalt_innskudd
    renteandel = (renteinntekter / sluttbelop) * 100
    return renteinntekter, renteandel


def vis_graf(saldo_over_tid, antall_ar, antall_innskudd):
    tidsakse = [i/antall_innskudd for i in range(antall_ar * antall_innskudd)]
    plt.plot(tidsakse, saldo_over_tid)
    plt.title("Utviklingen av sparekontoen")
    plt.xlabel("Tid (år)")
    plt.ylabel("Saldo (kr)")
    plt.show()


def main():
    # Input fra bruker
    terminbelop = float(input("Skriv inn terminbeløpet (beløpet du setter inn hver gang): "))
    rente = float(input("Skriv inn årlig rente (i prosent, f.eks 2 for 2%): ")) / 100
    antall_innskudd = int(input("Hvor mange innskudd per år? "))
    antall_ar = int(input("Hvor mange år skal du spare? "))

    
    totalt_innskudd = beregn_innskudd(terminbelop, antall_innskudd, antall_ar)

   
    sluttbelop, saldo_over_tid = beregn_sluttbelop(terminbelop, rente, antall_innskudd, antall_ar)

   
    renteinntekter, renteandel = beregn_renteandel(totalt_innskudd, sluttbelop)

    
    oppsummering(terminbelop, rente, antall_innskudd, antall_ar, sluttbelop, totalt_innskudd, renteinntekter, renteandel)


    vis_graf(saldo_over_tid, antall_ar, antall_innskudd)

main()
```
