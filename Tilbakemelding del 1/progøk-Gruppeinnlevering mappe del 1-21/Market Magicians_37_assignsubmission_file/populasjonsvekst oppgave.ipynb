```python
import math
def beregn_vekst(P, r, startar, sluttar, K):
    #A formell
    A = (K - P) / P
    #lagre variabler
    V = P
    Vmaks = 0
    ar_Vmaks = startar

    for t in range(startar, sluttar + 1):
        #regne ut populasjon ellerno
        Pt = K / (1 + A * math.exp(-r * (t - startar)))
        # sjekke maks pop
        if Pt > Vmaks:
            Vmaks = Pt
            ar_Vmaks = t
    return V, Vmaks, ar_Vmaks

# input
navn_omradet = input("Navn på området (f.eks. Island, Finnmark osv.)")
P = float(input("Nåværende populasjon (P)"))
r = float(input("Vekstrate (i prosent)")) / 100
startar = int(input("Startår"))
sluttar = int(input("Sluttår"))
K = float(input("Bæreevne (K)"))

#Utregning
V, Vmaks, ar_Vmaks = beregn_vekst(P, r, startar, sluttar, K)
#data bruker skriver inn
print(f"Område: {navn_omradet}")
print(f"Populasjonsvekst ved startår (V): {V}")
print(f"Maksimal populasjonsvekst (Vmaks): {Vmaks}")
print(f"År når maksimal populasjonsvekst oppnås (År Vmaks): {ar_Vmaks}")
```
