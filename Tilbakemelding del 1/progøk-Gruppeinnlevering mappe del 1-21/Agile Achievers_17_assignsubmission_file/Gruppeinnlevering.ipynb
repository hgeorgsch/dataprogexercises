# Sparekalkulator 1#

```python
import matplotlib.pyplot as plt

def regnut_sparing (P, r, n , t):
    F = P * (((1 + r / n) **(n * t) - 1) / (r/n))
    return F

def saldo_over_tid(P,r,n,t):
    saldo = []
    for aar in range(t + 1):
        F = regnut_sparing(P, r, n, aar)
        saldo.append(F)
    return saldo

#Tar inn bruker input
def main():
    P=float(input("Skriv ønsket terminbeløp"))
    r=float(input("Skriv rente"))
    n=int(input("Skriv hvor mange innskudd per år"))
    t = int(input("Skriv antall år"))

    siste_sum = regnut_sparing(P, r, n , t)

    total_innlegg= P * n * t
    total_rente= siste_sum - total_innlegg

#Oppsummering av kva du bruker har tastet inn
    print("\nOppsummering")
    print(f"Terminbeløp per periode: {P} ")
    print(f"Renten: {r*100} %")
    print(f"Antall innskuds perioder per år: {n}")
    print(f"Antall år: {t}")

#skriver ut resultatet av utregningene
    print("\nResultat")
    print(f"Sum i sparekonto: {siste_sum: .2f}")
    print(f"Totalt innskudd: {total_innlegg: .2f}")
    print(f"Totale renteavregninger: {total_rente: .2f}")
    print(f"Total rente i prosent: {(total_rente/total_innlegg)*100: .2f}%")


    aar = list(range(t + 1))
    saldo= saldo_over_tid(P, r, n, t)
      
#plotting av grafen   
    plt.plot(aar, saldo, label="Saldo over tid")
    plt.xlabel("År")
    plt.ylabel("Saldo (Kr)")
    plt.title("Utvikling av sparekonto over tid")
    plt.grid(True)
    plt.legend()
    plt.show()
    

if __name__== "__main__":
    main()
```

# Sparekalkulator 2#

```python
def sparekalkulator(
    initial_amount, 
    interest_rate, 
    inflation_rate, 
    wealth_tax_rate, 
    wealth_tax_threshold, 
    interest_tax_rate, 
    years):
    
    balance = initial_amount
    total_interest_earned = 0

    for year in range(1, years + 1):
        # Beregn årlig rente
        interest = balance * interest_rate
        total_interest_earned += interest
        
        # Skatt på renter
        tax_on_interest = interest * interest_tax_rate
        net_interest = interest - tax_on_interest
        
        # Legg til renter etter skatt
        balance += net_interest

        # Beregn formuesskatt hvis formuen overstiger terskelen
        if balance > wealth_tax_threshold:
            wealth_tax = (balance - wealth_tax_threshold) * wealth_tax_rate
            balance -= wealth_tax
        else:
            wealth_tax = 0

        # Juster for inflasjon
        inflation_adjustment = balance * inflation_rate
        balance -= inflation_adjustment

        # Print status for hvert år
        print(f"År {year}:")
        print(f"  Saldo før inflasjon: {balance + inflation_adjustment:.2f}")
        print(f"  Formuesskatt: {wealth_tax:.2f}")
        print(f"  Justering for inflasjon: {inflation_adjustment:.2f}")
        print(f"  Saldo etter inflasjon: {balance:.2f}")
        print(f"  Akkumulerte renter etter skatt: {total_interest_earned - (total_interest_earned * interest_tax_rate):.2f}")
        print("-" * 30)

    return balance


# Eksempel på bruk av sparekalkulatoren:
initial_amount = 1000000  # Startkapital
interest_rate = 0.05  # Årlig rente på 5%
inflation_rate = 0.02  # Årlig inflasjon på 2%
wealth_tax_rate = 0.01  # Formuesskatt på 1%
wealth_tax_threshold = 1500000  # Formuesskatt gjelder kun beløp over 1,5 mill
interest_tax_rate = 0.22  # Skatt på renteinntekter (22%)
years = 10  # Antall år

sluttbelop = sparekalkulator(
    initial_amount, 
    interest_rate, 
    inflation_rate, 
    wealth_tax_rate, 
    wealth_tax_threshold, 
    interest_tax_rate, 
    years)

print(f"Sluttbeløp etter {years} år: {sluttbelop:.2f} NOK")
```

# Populasjonsvekst

```python

import math
#parameterene
P0=3e6
K=8e6
r= 0.02
e=math.exp(1)
start_aar= 2024
slutt_aar= 2048
t=slutt_aar-start_aar
A = (K-P0)/P0

print(f"I start året 2024 ligger populasjonen til Litauen på {P0}")


#slutt populasjon
slutt_populasjon = K/(1+A*e**(-r*t))
print(f"Enden av 2048(aka sluttåret) ligger populasjonen til Litauen på {round(slutt_populasjon)}")

#økning i prosent
prosentvis_auke= (slutt_populasjon-P0)/P0*100
print(f"Populasjonen har gjennom dei 24 åra økt med {round(prosentvis_auke)} %")

#Tall på populasjon økning
pop_auke= slutt_populasjon-P0 
print(f"Det vil da si at populasjonen har økt med {round(pop_auke)} personer")
```

# Simulasjon av markedsdynamikk

```python
# Importerer nødvendige biblioteker
import numpy as np
import matplotlib.pyplot as plt

# Konstanter for etterspørsel og tilbud koeffisienter
a, b, c, d = 100, 1, 30, 2  # a og b for etterspørsel, c og d for tilbud
num_periods = 50  # Antall perioder for simulering
mean_shock, std_dev_shock = 0, 1  # Sjokkparametere for normalfordelte sjokk

# Lister for å lagre resultater for priser og kvantum
prices, quantities = [], []

# Simulerer markedsdynamikk over tid
for period in range(num_periods):
    # Genererer sjokk for etterspørsel og tilbud basert på normalfordeling
    shock_demand = np.random.normal(mean_shock, std_dev_shock)
    shock_supply = np.random.normal(mean_shock, std_dev_shock)

    # Beregner etterspørsel og tilbud med justering for sjokk
    demand = a - b * (c + d * (period + shock_demand))
    supply = c + d * (c + d * (period + shock_supply))

    # Beregner likevektspris som gjennomsnittet av etterspørsel og tilbud
    price = (demand + supply) / 2

    # Beregner likevektskvantum som minimum av etterspørsel og tilbud
    quantity = min(demand, supply)

    # Lagrer resultatene i lister
    prices.append(price)
    quantities.append(quantity)

    # Simulerer inngripen (for eksempel skatt) halvveis i simuleringen
    if period == num_periods // 2:
        b += 1  # Øker etterspørselskoeffisienten for å simulere en form for skatt

# Plotter resultatene
plt.figure(figsize=(12, 6))

# Plotter pris og kvantum over tid som linjer
plt.plot(prices, label='Likevektspris', color='blue', marker='o')
plt.plot(quantities, label='Likevektskvantum', color='orange', marker='o')

# Legger til tittel, akseetiketter og rutenett
plt.title('Likevektspris og Kvantum over tid')
plt.xlabel('Perioder')
plt.ylabel('Verdi')
plt.legend()
plt.grid()

# Optimaliserer layout og viser plottet
plt.tight_layout()
plt.show()
```
