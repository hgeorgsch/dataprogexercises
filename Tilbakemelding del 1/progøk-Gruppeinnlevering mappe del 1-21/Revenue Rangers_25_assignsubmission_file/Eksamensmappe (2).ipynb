# Populasjonsvekst:
Modellering med den logistiske modellen.

I denne oppgaven vil jeg be "brukeren" om å oppgi:
* Et land
* Vekstraten til landet
* Start år
* Slutt år
* Populasjon ved start år
* Makspopulasjon

Med disse tallene vil jeg regne ut:
* Sluttpopulasjonen i landet
* Populasjonens absolutte vekst
* Populasjonens relative vekst (i prosent)

Avslutningsvis fremvises populasjonsveksten fra start år til slutt år i et diagram, ut i fra de oppgitte tallene jeg fikk.

```python
land_fra_bruker = input("Oppgi ønsket land:")
vekstrate_i_landet = float(input("Hva er vekstraten i landet?")) #må legges inn med komma, ikke %
start_aar = int(input("Hva er start året?"))
slutt_aar = int(input("Hva er slutt året?"))
populasjon_start_aar = int(input("Hva er populasjonen ved start år?"))
makspopulasjon = int(input("Hva er makspopulasjonen?"))

P0 = populasjon_start_aar
r = vekstrate_i_landet
K = makspopulasjon
t = slutt_aar - start_aar
e = 2.71828
A = (K-P0)/P0

#Beregning av populasjon over tid:
aar = list(range(start_aar, slutt_aar + 1))
populasjoner = []

for t in range(len(aar)):
    populasjon = K/(1+A*e**(-r*t)) #den logistiske vekstmodellen
    populasjoner.append(populasjon)

#Beregning av sluttpopulasjon, absolutt vekst og relativ vekst:
sluttpopulasjon = populasjoner[-1]
absolutt_vekst = sluttpopulasjon - P0
relativ_vekst = (absolutt_vekst / P0) * 100

print("\n""Du har gitt meg landet", land_fra_bruker,
      "\n""med vekstraten", vekstrate_i_landet,
      "\n""og en makspopulasjon på", makspopulasjon, end="\n\n")

print("Sluttpopulasjonen i landet er:", round(sluttpopulasjon), "mennesker")
print("Sluttpopulasjonen avrundet til nærmeste hele ti tusen er:", round(sluttpopulasjon, -4),end="\n\n")

print("Populasjonens absolutte vekst er på", round(absolutt_vekst), "personer")
print("Populasjonens realtive vekst er", round(relativ_vekst, +2),"%",end="\n\n")

plt.figure(figsize=(6,4))
plt.bar(aar, populasjoner, label="Populasjon")
plt.legend()
plt.grid()
plt.title(f"(Absolutt) Populasjonsvekst i {land_fra_bruker} ({start_aar}-{slutt_aar})")
plt.xlabel("År")
plt.ylabel("Populasjon (i millioner)")
plt.show()
```

# Forklaring av resultater - populasjonsvekst:
Denne koden beregner populasjonsutviklingen i et angitt land basert på den logistiske vekstmodellen, denne tar hensyn til en øvre grense for hvor stor populasjonen kan bli (makspopulasjon). Resultatene gir en oversikt over populasjonsendringer over tid fra et angitt startår – sluttår.

På grunn av måten denne koden er skrevet på, vil den oppgi forskjellige resultater basert på materialet som blir oppgitt, vi vil da kunne sjekke forskjellige data med samme kode.

* Sluttpopulasjon: beregner den estimerte populasjonen ved sluttår, dette representerer antall mennesker som bor i landet ved periodens slutt, ut ifra vekstraten og makspopulasjonen som er oppgitt.

* Absolutt vekst: viser differansen mellom populasjonen ved periodens slutt og populasjonen ved startåret. 

* Relativ vekst: populasjonsveksten i prosentandel ved startår. Dette gir en indikasjon på hvor stor økningen har vært i forhold til den opprinnelige populasjonen.

* Fremstilling av populasjonsvekst (diagram): stolpediagrammet som er generert viser populasjonen for hvert år i den oppgitte tidsperioden. Denne visualiserer hvordan populasjonen endrer seg over tid med en tydelig vekst frem til nær makspopulasjon, der vekstraten vil avta.

Ved å bruke den logistiske vekstmodellen, tar koden hensyn til det faktum at veksten vanligvis avtar når populasjonen nærmer seg makspopulasjonsgrensen på grunn av begrensede ressurser.

En slik type kode for populasjonsvekst kan vi bruke og tilpasse i mange forskjellige sammenhenger hvor logistisk vekst er relevant. For eksempel: økonomisk vekst, sykdomsspredning, bakterievekst, vektøkning/vektrmyejon og mange mer.


# Sparekalkulator:

I denne oppgaven vil jeg be "brukeren" om å oppgi:
* Terminbeløp per mnd
* Rentesats
* Antall innskudd per år
* Hvor mange år det spares

Med disse tallene vil jeg regne ut:
* Sluttbeløpet på sparekontoen
* Innskudd i kroner og %
* Renter i kroner og %

Avslutningsvis vil jeg fremvise skatt, formueskatt og inflasjon i et søylediagram, og innskudd og renter i et kakediagram.

```python
terminbelop = int(input("Terminbeløpet per mnd:"))
rente = float(input("Rentesats:"))
innskudd = int(input("Antall innskudd per år:"))
aar = int(input("Hvor mange år spares det?"))

inflasjon = 0.02 # 2% inflasjon
formueskatt = 0.02 # 2% formueskatt
skatteprosent = 0.22 # 22% skatt på renteinntekter

P = terminbelop
r = rente
n = innskudd #per år
t = aar #antall år

#Lister for prosentandeler, skatt, formueskatt og inflasjon:
innskudd_prosent_liste = []
renter_prosent_liste = []
skatt_liste = []
formueskatt_liste = []
inflasjon_liste = []

#Beregner prosentandeler for hvert år:
for år in range(1, t+1):
   
    #beregner sluttbeløpet:
    slutt_belop = P * (((1 + (r / n))**(n * t)-1)/(r / n))
    
    #beregner totalt innskudd (uten renter) og rentebeløpet:
    totalt_innskudd = P * n * t
    rente_belop = slutt_belop - totalt_innskudd

    #rentebeløp for skatt, formueskatt og inflasjon:
    skatt_renter = rente_belop * skatteprosent
    formueskatt_belop = slutt_belop * formueskatt
    inflasjon_belop = slutt_belop * inflasjon
    
    #justerer rentebeløpet:
    justert_rente_belop = rente_belop - skatt_renter - (slutt_belop * formueskatt) - (slutt_belop * inflasjon)

    #beregner prosentfordelingen (basert på justerte rentebeløp):
    totalt_slutt_belop = totalt_innskudd + justert_rente_belop
    innskudd_prosent = (totalt_innskudd / totalt_slutt_belop) * 100
    renter_prosent = (justert_rente_belop/totalt_slutt_belop) * 100

    #Legger til verdiene i listene:
    innskudd_prosent_liste.append(innskudd_prosent)
    renter_prosent_liste.append(renter_prosent)
    skatt_liste.append(skatt_renter)
    formueskatt_liste.append(formueskatt_belop)
    inflasjon_liste.append(inflasjon_belop)

slutt_belop_avrundet = int(round(totalt_slutt_belop))
totalt_innskudd_avrundet = int(round(totalt_innskudd))
rente_belop_avrundet = float(round(justert_rente_belop))

print("\n""Sluttbeløpet på sparekontoen er", slutt_belop_avrundet, "etter", t, "år")
print("Av sluttbeløpet er", totalt_innskudd_avrundet, "kr (", round(innskudd_prosent, 2), "%) innskudd,")
print("og", rente_belop_avrundet, "kr (", round(renter_prosent, 2), "%) renter.")

#Visulaisering av skatt, formueskatt og inflasjon i et søylediagram:
År = np.arange(1, t+1)
bar_width = 0.2
plt.figure(figsize=(6, 4))
plt.bar(År - bar_width / 2, skatt_liste, bar_width, label='Skatt', color='red')
plt.bar(År, formueskatt_liste, bar_width, label='Formueskatt', color='yellow')
plt.bar(År + bar_width / 2, inflasjon_liste, bar_width, label='Inflasjon', color='green')
plt.xlabel("År")
plt.ylabel("Skatt, formueskatt og inflasjon over tid")
plt.grid()
plt.legend()
plt.show()

#Visualisering av renter og innskudd i et kakediagram:
labels = ['Innskudd', 'Renter']
sizes = [innskudd_prosent_liste[-1], renter_prosent_liste[-1]]
colors = ['orange', 'blue']

plt.figure(figsize=(6, 4))
plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=140)
plt.title("Fordeling av innskudd og renter")
plt.show()
```

# Forklaring av resultater - sparekalkulator:
Denne koden gir en klar visualisering av hvordan ulike kostnader påvirker det totale beløpet på sparekontoen, og viser fordelingen mellom renter og innskudd.

* Sluttbeløp på sparekontoen: denne viser totalbeløpet på konto etter oppgitte antall år, og inkluderer både innskudd og renter.

# Simuleringsprogram:

Simulering med den logistiske modellen.
Vi ønsker å undersøke hvordan det å sette varer på tilbud påvirker salg og omsetning.

* Anta at en butikk har et knippe produkter til en viss pris.
* Når en kunde kommer inn i butikken er det en viss sjanse knyttet til hvert produkt for at kunden kjøper det, men dersom varene kommer på salg øker denne sjansen.

I denne oppgaven:
* Gjør jeg noen antagelser om hvor mange kunder som kommer innom butikken hver dag.
* Når en kunde kommer inn i butikken har de et visst budjsett som trekkes tilfeldig fra en passende fordeling
* Kunden ser så gjennom produktene og kjøper de basert på tilfeldighet (produktets parametre) og om de har penger igjen i budsjettet.

* Videre sammenlignes salg og omsetning for en måned med og uten produkter på tilbud.

tabell (pandas)

```python
#Produkter med pris og kjøpssjanse uten tilbud:
produkter = {
    'hårprodukt_1': {'pris': 200, 'kjopesjanse': 0.3},
    'hårprodukt_2': {'pris': 350, 'kjopesjanse': 0.15},
    'hårprodukt_3': {'pris': 500, 'kjopesjanse': 0.2},
    'hårprodukt_4': {'pris': 250, 'kjopesjanse': 0.1},
    'hårprodukt_5': {'pris': 400, 'kjopesjanse': 0.25}
}

#Funksjon for å simulere kunde med et tilfeldig budsjett
def generer_budsjett():
    return random.gauss(500, 150)  # Normalfordeling rundt 500 (middelverdi) med standardavvik 150

#Funksjon for å simulere sjansen for kjøp med/uten tilbud:
def kjopesjanse(med_tilbud, grunn_sjanse):
    if med_tilbud:
        return 1 / (1 + np.exp(-5 * (grunn_sjanse + 0.3)))  #logistisk modell
    else:
        return grunn_sjanse

#Funksjon for å simulere kunde som går gjennom butikken:
def simuler_kunde(med_tilbud):
    budsjett = generer_budsjett()
    total_kjop = 0
    omsetning = 0
    
    for produkt, param in produkter.items():
        pris = param['pris']
        sjanse = kjopesjanse(med_tilbud, param['kjopesjanse'])
        
        #Kunden kjøper produktet dersom sjansen er stor nok og at de har nok penger
        if random.random() < sjanse and budsjett >= pris:
            total_kjop += 1
            omsetning += pris
            budsjett -= pris

    return total_kjop, omsetning

#Simulerer en dag med kunder:
def simuler_dag(antall_kunder, med_tilbud):
    total_salg = 0
    total_omsetning = 0
    
    for _ in range(antall_kunder):
        kjop, omsetning = simuler_kunde(med_tilbud)
        total_salg += kjop
        total_omsetning += omsetning
    
    return total_salg, total_omsetning

# Eksempel på simulering for 1 dag med 250 kunder
antall_kunder = 250
salg_uten_tilbud, omsetning_uten_tilbud = simuler_dag(antall_kunder, med_tilbud=False)
salg_med_tilbud, omsetning_med_tilbud = simuler_dag(antall_kunder, med_tilbud=True)

print(f"Salg uten tilbud: {salg_uten_tilbud}, Omsetning uten tilbud: {omsetning_uten_tilbud}")
print(f"Salg med tilbud: {salg_med_tilbud}, Omsetning med tilbud: {omsetning_med_tilbud}")
```

I denne koden vil det "simuleres tilfeldig", sluttresultatet (salg uten/med tilbud) vil derfor variere for hver gang.
