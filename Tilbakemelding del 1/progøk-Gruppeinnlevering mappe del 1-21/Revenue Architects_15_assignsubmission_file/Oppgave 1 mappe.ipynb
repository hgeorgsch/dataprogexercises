```python
import math

def logistisk_vekst(P0, K, r, t):
    A = (K - P0) / P0
    P = K / (1 + A * math.exp(-r * t))
    return P


def populasjonsvekst_program(navn_område, vekstrate, startår, sluttår, populasjon_start, maks_populasjon):

    år = sluttår - startår
    

    populasjon_slutt = logistisk_vekst(populasjon_start, maks_populasjon, vekstrate, år)
    

    absolutt_vekst = populasjon_slutt - populasjon_start
    relativ_vekst = (absolutt_vekst / populasjon_start) * 100
    

    print(f"Oppsummering for {navn_område}:")
    print(f"Vekstrate: {vekstrate*100}%")
    print(f"Startår: {startår}")
    print(f"Sluttår: {sluttår}")
    print(f"Populasjon ved startår: {int(populasjon_start)}")
    print(f"Maksimal populasjon: {int(maks_populasjon)}")
    print(f"Populasjon ved sluttår: {int(populasjon_slutt)}")
    print(f"Absolutt vekst: {int(absolutt_vekst)} personer")
    print(f"Relativ vekst: {relativ_vekst:.2f}%")
    

navn_område = "Island"
vekstrate = 0.02  
startår = 2000
sluttår = 2100
populasjon_start = 400000  
maks_populasjon = 1000000


populasjonsvekst_program(navn_område, vekstrate, startår, sluttår, populasjon_start, maks_populasjon)
```

```python
import matplotlib.pyplot as plt

# Data for Island (vekst over 100 år fra 2000 til 2100)
navn_område = "Island"
vekstrate = 0.02  # 2% vekstrate
startår = 2000
sluttår = 2100
populasjon_start = 400000  # Populasjon ved startår
maks_populasjon = 1000000  # Maksimal populasjon


def logistisk_vekst(P0, K, r, t):
    A = (K - P0) / P0
    P = K / (1 + A * math.exp(-r * t))
    return P


år_liste = list(range(sluttår - startår + 1))  
populasjon_liste = [logistisk_vekst(populasjon_start, maks_populasjon, vekstrate, år) for år in år_liste]


plt.plot(år_liste, populasjon_liste, label="Populasjon på Island", color='green')

plt.title(f"Populasjonsvekst i {navn_område} fra {startår} til {sluttår}")
plt.xlabel("År")
plt.ylabel("Populasjon")
plt.legend()

# Vis grafen
plt.show()
```
