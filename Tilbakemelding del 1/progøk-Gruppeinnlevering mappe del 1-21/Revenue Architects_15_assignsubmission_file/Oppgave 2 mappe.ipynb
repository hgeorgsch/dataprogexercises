```python
import math

def finn_sluttbeløp(P, r, n, t):
    rn = r / n
    F = P * (((1 + rn) ** (n * t) - 1) / rn)
    return F

def spareprogram():
    P = float(input("Skriv inn terminbeløpet (P): "))
    r = float(input("Skriv inn årlig rente i desimalform (f.eks. 0,02 for 2%): "))
    n = int(input("Skriv inn antall innskudd per år (n): "))
    t = int(input("Skriv inn antall år (t): "))

    sluttbeløp = finn_sluttbeløp(P, r, n, t)

    total_innskudd = P * n * t

    renter = sluttbeløp - total_innskudd
    
    print(f"\nOppsummering:")
    print(f"Terminbeløp: {P} kr")
    print(f"Rente: {r*100}%")
    print(f"Antall innskudd per år: {n}")
    print(f"Antall år: {t}")
    print(f"\nSluttbeløp på sparekonto: {sluttbeløp:.2f} kr")
    print(f"Total innskudd: {total_innskudd:.2f} kr")
    print(f"Renter: {renter:.2f} kr")
    print(f"Renter i prosent av sluttbeløp: {renter / sluttbeløp * 100:.2f}%")

spareprogram()
```
