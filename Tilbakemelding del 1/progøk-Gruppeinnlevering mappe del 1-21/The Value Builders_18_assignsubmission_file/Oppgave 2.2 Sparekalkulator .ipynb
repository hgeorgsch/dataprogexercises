# Sparekalkulator 
Som vist nedenfor har jeg laget en sparekalkulator. Jeg vil finne ut hvor mye du kan oppnå ved å spare penger over tid.
Det hjelper å finne ut hvor stort sparebeløpet vil være etter et bestemt antall år, inkludert rentene. Hvis man endrer 
startbeløp, rentesatsen, eller spareperioden kan man se at det påvirker beløpet som blir spart.  

Alt dette gjøres ved å oppgi et startbeløp hvor rentene blir lagt til, altså den årlige renten og hvor mange år fremmover 
det er planlagt å spare. Ut ifra opplysningene regner phyton 3 ut hvor mye penger man vil ha til slutt, både fra det man 
har spart og tjent opp i renter. Informasjonen viser også prosentvis fordeling mellom innskudd og renter. 

```python
# Koden for oppgave 2.2

import matplotlib.pyplot as plt

def sparing_beregning():

    startbeløp = int(input("Startbeløp: "))
    antall_renteberegninger = int(input("Antall ganger i året renter legges til: "))
    årlig_rente = float(input("Årlig rente (i desimaltall, f.eks. 0.05 for 5%): "))
    spareperiode = int(input("Sparingsperiode (i år): "))

    
    sluttbeløp = startbeløp * (((1 + årlig_rente / antall_renteberegninger) ** (antall_renteberegninger * spareperiode) - 1) / (årlig_rente / antall_renteberegninger))
    avrundet_beløp = round(sluttbeløp)


    totalt_spart = startbeløp * spareperiode * antall_renteberegninger
    opptjente_renter = avrundet_beløp - totalt_spart


    spart_prosent = round((totalt_spart * 100) / avrundet_beløp)
    renter_prosent = round((opptjente_renter * 100) / avrundet_beløp)

    return startbeløp, antall_renteberegninger, årlig_rente, spareperiode, totalt_spart, avrundet_beløp, opptjente_renter, spart_prosent, renter_prosent


startbeløp, antall_renteberegninger, årlig_rente, spareperiode, totalt_spart, sluttbeløp, opptjente_renter, spart_prosent, renter_prosent = sparing_beregning()


år_liste = list(range(0, spareperiode + 1))
sluttbeløp_liste = [startbeløp * (((1 + årlig_rente / antall_renteberegninger) ** (antall_renteberegninger * år) - 1) / (årlig_rente / antall_renteberegninger)) for år in range(spareperiode + 1)]
spart_liste = [startbeløp * antall_renteberegninger * år for år in range(spareperiode + 1)]
renter_liste = [sluttbeløp_liste[år] - spart_liste[år] for år in range(spareperiode + 1)]


print("Sluttbeløp:", sluttbeløp)
print("Totalt spart: kr", totalt_spart, "altså", spart_prosent, "%")
print("Opptjente renter: kr", opptjente_renter, "altså", renter_prosent, "%")


plt.plot(år_liste, sluttbeløp_liste, label="Sluttbeløp", color='green', linewidth=2)
plt.plot(år_liste, renter_liste, label="Opptjente renter", color='blue', linestyle='--')
plt.plot(år_liste, spart_liste, label="Totalt spart", color='red', linestyle='-.')
plt.xlabel("År")
plt.ylabel("Beløp (kr)")
plt.legend()  
plt.grid(True)  
plt.show()
```

# Resultater 
Grafen ovenfor viser følgende, et startbeløp på 10 000 kr, og en årlig rente på 3%. Renten legges til årlig over en sparingsperiode på 15 år, da endte man opp med et totalbeløp på 2 269 727kr. Av dette beløpet kommer 1 800 000 fra egne innskudd, som utgjør 
i prosent 79% av den totale sluttsummen. Opptjente renter ble 469 727 kr, som utgjør i prosent 21% av den totale sluttsummen.
