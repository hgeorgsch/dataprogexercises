Logisk lik, ` == `, er enten sant eller usant, mens `=` er en tilordningsoperator, og har ikke noe sannhetsverdi. 


Vi utvider programmet til å også gi tilbakemelding ved andre karakterer. Da bruker vi nøkkelordet `else`.  

```python
karakter = input("hvilken karakter fikk du på mappa? ")
if karakter == 'A' :
    print ("Du må ha jobbet mye")
else:
    print("Det viktigste er å lære, ikke karakteren")
```

# Betingelser

* Konstrueres med _sammenligningsoperatorer_.

|Operator | Betydning|
|---|---|
|> | Større enn|
|>= | Større enn eller lik|
|< | Mindre enn|
|<= | Mindre enn eller lik|
|== | Er lik|
|!= | Er ulik|

* Resultatet av en utregning med bruk av en _sammenligningsoperator_ er enten `True` eller `False`.
* NB! Pass på at du ikke blander _tilordning_ `a=5` med _evaluering_ `a==5`.

La oss se hvordan vi kan bruke dette for å gi tilbakemelding om karakter på videregående:

```python
karaktersnitt =float(input("Hei, hva fikk du i karaktersnitt på vitnemålet? "))
if karaktersnitt > 2.5:
    print(karaktersnitt, "er bedre enn meg. Nå ble jeg sur")
else:
    karaktersnitt <= 2.5
    print(karaktersnitt, "er dårligere enn meg. Nå ble jeg glad :) ")
```

Ofte er det flere enn to mulige alternativer. Da kan det være nyttig med `elif`.  


```python
karaktersnitt =float(input("Hei, hva fikk du i karaktersnitt på vitnemålet? "))
if karaktersnitt > 2.5:
    print(karaktersnitt, "er bedre enn meg. Nå ble jeg sur")
elif karaktersnitt < 0.0:
    print("Negativt karaktersnitt, det er det første gang jeg hører om.")
elif karaktersnitt == 2.5:
    print("Samme som meg. Vi er forbundet med en usynlig tråd")
else:
    karaktersnitt <= 2.5
    print(karaktersnitt, "er dårligere enn meg. Nå ble jeg glad :) ")
```

## "og" og "eller" på dataspråk

Det kan også være nyttig å kombinere flere logiske vilkår til større uttrykk, fordi vi har flere vilkår som skal oppfylles. Da er det nyttig med logiske operatorer.   
Operator | Beskrivelse
---|---
`or` | Logisk eller (OR)
`and` | Logisk og (AND)
`not` | Logisk ikke, negasjon (NOT)

For å illustrere hvordan vi kan bruke funksjoner i programmering, og at det gjør programmene våre mer generelle, lager vi en python funksjon som kan brukes uavhengig av hvilken karakter vi selv fikk på vitnemålet:

```python
def svar_karakter(min_karakter):
    karaktersnitt =float(input("Hei, hva fikk du i karaktersnitt på vitnemålet? "))
    if karaktersnitt > min_karakter and karaktersnitt < 6.0:
        print(karaktersnitt, "er bedre enn meg. Nå ble jeg sur")
    elif karaktersnitt < 0.0:
        print("Negativt karaktersnitt, det er det første gang jeg hører om.")
    elif karaktersnitt > 6.0:
        print("Mer enn 6.0, du er meg nå en god skrytepave")
    elif karaktersnitt == min_karakter:
        print("Samme som meg. Vi er forbundet med en usynlig tråd")
    else:
        karaktersnitt <= min_karakter
        print(karaktersnitt, "er dårligere enn meg. Nå ble jeg glad :) ")
    
```

```python
svar_karakter(5.1)
```

```python
import math
# Definer alle parametre som variabler
e = math.exp(1) # = e^1
P0_norge = 5e6 # Startpopulasjon Norge
P0_sverige = 9e6 # Startpopulasjon Sverige
K_norge = 59e6
K_sverige = 25e6 # Makspopulasjon Sverige
r_norge = 0.02
r_sverige = 0.01 #Vekstrate Sverige
```

```python
def regnutA(P0,K):
    A = (K-P0)/P0
    return A

def beregn_populasjon_norge(t):
    A_norge=  regnutA(K_norge,P0_norge)
    P_slutt = K_norge/(1+A_norge*e**(-r_norge*t))
    return P_slutt

def beregn_populasjon_sverige(t):
    A_sverige = regnutA(K_sverige,P0_sverige)
    P_slutt = K_sverige/(1+A_sverige*e**(-r_sverige*t))
    return P_slutt
def skriv_ut_populasjon_sve_nor(t):
    norge = beregn_populasjon_norge(t)
    sverige = beregn_populasjon_sverige(t)
    if t <0:
        print("Vi ser på fortiden")
    if norge > sverige:
        print("I følge modellen vil det være ",int(norge - sverige),
              " flere innbygger i Norge enn i Sverige i ", int(2025+t))
    else:
        print("I følge modellen vil det være ",int(sverige - norge),
              " flere innbygger i Sverige enn i Norge i ", int(2025+t))
            
skriv_ut_populasjon_sve_nor(-20)
```
