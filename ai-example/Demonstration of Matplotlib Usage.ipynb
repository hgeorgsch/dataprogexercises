## Basic Plotting with Matplotlib

 ```python
# Import necessary libraries
import matplotlib.pyplot as plt
import numpy as np

````python
# Basic Plotting with Matplotlib
# In this section, we will learn how to create basic plots using Matplotlib in Python.
````

````python
# Step 1: Create a Line Plot
x = np.linspace(0, 10, 100)
y = np.sin(x)
````

````python
plt.figure(figsize=(8, 6))  # Set the figure size
plt.plot(x, y, label='Sine Wave', color='blue')  # Plot the line plot
plt.title('Line Plot of Sine Function')  # Add a title to the plot
plt.xlabel('X-axis')  # Label for the x-axis
plt.ylabel('Y-axis')  # Label for the y-axis
plt.legend()  # Show the legend
plt.grid(True)  # Add grid lines to the plot
plt.savefig('line_plot.png')  # Save the plot as an image file
plt.show()  # Display the plot
````

````python
# Step 2: Create a Scatter Plot
plt.figure(figsize=(8, 6))  # Set the figure size
plt.scatter(x, y, color='red', label='Scatter Points')  # Plot the scatter plot
plt.title('Scatter Plot of Sine Function')  # Add a title to the plot
plt.xlabel('X-axis')  # Label for the x-axis
plt.ylabel('Y-axis')  # Label for the y-axis
plt.legend()  # Show the legend
plt.grid(True)  # Add grid lines to the plot
plt.savefig('scatter_plot.png')  # Save the scatter plot as an image file
plt.show()  # Display the plot
````

````python
# Step 3: Create a Bar Chart
categories = ['A', 'B', 'C', 'D']
values = [10, 15, 12, 8]
````

````python
plt.figure(figsize=(8, 6))  # Set the figure size
plt.bar(categories, values, color='green')  # Plot the bar chart
plt.title('Bar Chart of Categories vs Values')  # Add a title to the plot
plt.xlabel('Categories')  # Label for the x-axis
plt.ylabel('Values')  # Label for the y-axis
plt.grid(True)  # Add grid lines to the plot
plt.savefig('bar_chart.png')  # Save the bar chart as an image file
plt.show()  # Display the plot
````

````python
# Step 4: Create a Histogram
data = np.random.normal(size=1000)
````

````python
plt.figure(figsize=(8, 6))  # Set the figure size
plt.hist(data, bins=30, color='purple', edgecolor='black')  # Plot the histogram
plt.title('Histogram of Random Data')  # Add a title to the plot
plt.xlabel('Value')  # Label for the x-axis
plt.ylabel('Frequency')  # Label for the y-axis
plt.grid(True)  # Add grid lines to the plot
plt.savefig('histogram.png')  # Save the histogram as an image file
plt.show()  # Display the plot
````

## Subplots in Matplotlib

 ```python
# Import necessary libraries
import matplotlib.pyplot as plt
import numpy as np

````python
# Create a figure and a set of subplots
fig, axs = plt.subplots(2, 2)  # Creates a 2x2 grid of subplots
````

````python
# Generate some data to plot
x = np.linspace(0, 2 * np.pi, 400)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)
y4 = np.exp(np.sin(x))
````

````python
# Plot the data on each subplot
axs[0, 0].plot(x, y1, 'r')       # First subplot: sine wave
axs[0, 0].set_title('Sine Wave')
````

````python
axs[0, 1].plot(x, y2, 'g')      # Second subplot: cosine wave
axs[0, 1].set_title('Cosine Wave')
````

````python
axs[1, 0].plot(x, y3, 'b')      # Third subplot: tangent wave
axs[1, 0].set_title('Tangent Wave')
````

````python
axs[1, 1].plot(x, y4, 'y')     # Fourth subplot: exponential sine wave
axs[1, 1].set_title('Exponential Sine Wave')
````

````python
# Adjust layout and display the plots
plt.tight_layout()  # Automatically adjusts subplots to give some padding and make room for titles
plt.show()           # Display the figure with all subplots
````
