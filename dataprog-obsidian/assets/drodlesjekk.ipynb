```python
import matplotlib.pyplot as plt
import math
import numpy as np

beta0 = 0.2
beta1 = 1
def Ps(disc,P0=0.2,r=5):
    K = 1
    A = (K-P0)/P0
    return K/(1+A*math.exp(-(r*disc)))

x = np.linspace(0,1,25)
print(x)
y = [Ps(xi) for xi in x]
plt.plot(x,y)
```
