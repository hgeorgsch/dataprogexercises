```python
import pandas as pd
import numpy as np
import datetime

df = pd.read_csv("befolkning.csv", encoding="ISO-8859-1", sep=";", header=1)
df = df.replace("..", np.nan)
intKols = list(df.columns)
intKols.remove("kvartal")
intKols.remove("region")
typedikt = {kol : "Int32" for kol in intKols}
df = df.astype(typedikt)
df["kvartal"] = df["kvartal"].map(lambda x: pd.Period(x.replace("K", "Q"), freq="Q"))
df = df.set_index(["region", "kvartal"])

def finn_kommune(kommune):
    for sted, kvartal in df.index:
        if kommune in sted:
            return sted
    return None

kode_aalesund = finn_kommune("Ålesund")
print(kode_aalesund)

df.loc[kode_aalesund, '1997Q4']["Utvandring"]
```
