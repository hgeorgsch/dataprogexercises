# Oppgave 4
Vi Ser på tilfellet fra oppgave 1:
$$
P(t) = \frac{K}{1+Ae^{-rt}}
$$
Her er:
* $P(t)$: Populasjon ved tiden t
* $t$ er tiden
* $r$ er relativ vekstrate, feks 2%
* $K$ er makskapasiteten til populasjonen
* $e$ er eulers tall: $e\approx 2.71828$

$A$ er en koeffisient definert ved:
$$
A = \frac{K-P_0}{P_0}
$$

Dere skal lage en utvidelse av programmet dere lager der dere bruker det dere har lært. 
Programmet spør bruker etter:
-  Navn på to ulike områder (eg. Island, Finmark osv)
-  Hvert lands vekstrate 
-  Hvert lands startpopulasjon
-  Hvert lands makskapasitet

Startår og sluttår er felles for de to landene og gitt ved 2023 og 2040. 

Programmet skal returnere en liste med de følgende elementene:
-  En streng med navnet til det området med høyest populasjon ved sluttår. 
-  Poplulasjonen til det landet ved sluttår
