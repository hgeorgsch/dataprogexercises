Funskjoner er nyttig fordi de løser oppgaver for oss. La oss se på hvordan vi kan løse noen av oppgave fra forrige notebook ved å bruke funksjoner. Først repeterer vi hvordan vi løste problemet uten å bruke funksjoner. Modellen er som følger: 

$$
P(t) = \frac{K}{1+Ae^{-rt}}
$$
Her er:
* $P(t)$: Populasjon ved tiden t
* $t$ er tiden
* $r$ er relativ vekstrate, feks 2%
* $K$ er makskapasiteten til populasjonen
* $e$ er eulers tall: $e\approx 2,71828$

```python
#Folketall på Island

e = 2.71828 #Eulers tall (Vi skal se en bedre måte å hente slike konstanter)
K = 4e6 #Maksimal populasjon
r = 0.02 #Relativ vekstrate
start = 2021 #Startår
slutt = 2040 #Sluttår
P0 = 372520 #Startpopulasjon

print("Ved år", start, "var innbyggertallet på Island", P0)
print("Vi antar en årlig vekstrate på", r*100, "%")

tid = slutt-start #Hvor lang tid mellom startår og sluttår
A = (K-P0)/P0   #Koeffisient fra formel

populasjon_2040 = int( K/(1+A*e**(-r*tid))) #Populasjon i år 2040 - som heltall

#Vi vil runde av til nærmeste 1000
#Vi kan gjøre det på 4 måter
populasjon_avrundet = round(populasjon_2040/1000,0)*1000
populasjon_avrundet = int(populasjon_avrundet)

populasjon_avrundet2 = populasjon_2040-populasjon_2040%1000

populasjon_avrundet3 = (populasjon_2040//1000)*1000

populasjon_avrundet4 = round(populasjon_2040,-3)

#print("test", populasjon_avrundet4)

#print(tid, "år senere, i", slutt, "vil populasjon være", populasjon_avrundet4)
print(tid, "år senere, i", slutt, "vil populasjon være omtrent", populasjon_2040//1000, "tusen", "\n \n")
```

# Hvorfor bruke funksjoner?
Eksemplene med innbyggertallet på Island og panikkalder lot seg fint løse uten funksjoner. Hvorfor er de nødvendige?
-  Koden blir mer lesbar.
-  En kan gjennbruke samme funksjon med forskjellige parametere flere ganger i løpet av et program. Dersom en skulle ha brukt klipp og lim, måtte en endre koden på mange steder dersom den måtte endres.
-  Organisering av arbeid. Dersom ditt bidrag i utviklingen er å lage en funsjon trenger du bare at person B gir deg de parameterene du trenger og du trenger bare  å sørge for at funksjonen returnerer det person C trenger.
-  Design av et program. Det er mye lettere å designe et program dersom man først bestemmer hva man trenger fra funksjonene, og senere tenker på hvordan funksjonene skal implementeres.
