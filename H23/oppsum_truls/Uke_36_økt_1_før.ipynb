La oss si at vi vil ta høyde for et tredje tilfelle, spesialtilfellet at de to tallene er like. Da skal funksjonen returnere 3*x. I python løser vi dette ved å bruke elif. 


```python
#Vi lager en funksjon som tar inn to tall og returnerer det største tallet multiplisert med 2.3

def double_largest(x,y):
    if(x>y):
        return 2*x
    elif(x==y):
        return 3*x
    else:
        return 2*y
print(double_largest(5,5))

```

Det er også mulig med flere elif:

```python
#Funksjonen double_largest aksepterer ikke at noen av leddene er lik 0. I så tilfelle returneres
#en feilmelding
def double_largest(x,y):
    if(x*y==0):
        return "zero values are not accepted"
    elif(x>y):
        return 2*x
    elif(x==y):
        return 3*x
    else:
        return 2*y
print(double_largest(6,6))
```

Vi skal nå se på hvordan vi kan bruker for-løkker til å lage en liste med forskjellen i kontobeholdning ved to forskjellige forrentninger etter hvert år, i et spesifisert antall år:

```python
def penge_forskjell(rente_1, rente_2,startsum, antall_år):
    forskjell_liste = []#Før vi fyller listen må vi lage en tom liste
    sum_1 = startsum
    sum_2 = startsum
    for i in range(antall_år):
        sum_1 = sum_1*rente_1
        sum_2 = sum_2*rente_2
        forskjell_liste.append(round(sum_2-sum_1,2))
    return forskjell_liste
forskjell_test= penge_forskjell(1.01, 1.05, 250000,10)
print(forskjell_test)

```

For en ny illustrajon av for løkker kan vi se på befolkningsmodellen vi har sett på før.  

```python

e = 2.71828 #Eulers tall (Vi skal se en bedre måte å hente slike konstanter)
K = 4e6 #Maksimal populasjon
r = 0.02 #Relativ vekstrat
start = 2021 #Startår
slutt = 2040 #Sluttår
P0 = 372520 #Startpopulasjon

print("Ved år", start, "var innbyggertallet på Island", P0)
print("Vi antar en årlig vekstrate på", r*100, "%")

tid = slutt-start #Hvor lang tid mellom startår og sluttår
A = (K-P0)/P0   #Koeffisient fra formel
pop_liste = []
temp_tid = 0#tiden settes lik starttiden
for i in range(tid):
    
    pop_liste.append(int( K/(1+A*e**(-r*temp_tid))))
    temp_tid = temp_tid + i
    
print(pop_liste) 
my_sum = 0
for i in range(tid):
    print(i)
    my_sum = my_sum + pop_liste[i]
    
my_sum = my_sum/tid
print(my_sum)
```
