

## Funksjoner
I tilleg til pluss, minus, gange og deling, er det praktisk å kunne bruke funksjoner 
som gjør litt andre og ofte mer kompliserte ting. I python finnes
- innebyggde funksjoner
- egendefinerte funksjoner

Innebygde funksjoner er veldig praktiske, siden de er laget ferdig for oss. 

```python
# En innebygget funksjon vi har brukt mye hittil er print(). 
print("Hei verden")
#round() er et typisk eksempel på en innebygget småfunksjon
print(round(23234234,-4))
```

```python
# For å løse mer sammensatte oppgaver er vi ofte avhengig av å lage våre egne funksjoner. Disse 
# kan ofte bruke innebyggde funksjoner.
#Først må vi definere funksjonen:
def my_print_round(number, message):
    print("The message is: ", message +".", "The rounded number is:", round(number,1))
```

Til nå har vi kun definert funksjonen. For at datamaskinen skal utføre noe, må vi kalle funksjonen. 

```python
my_print_round(34.34, "Vær fri")
```

Denne funksjonen returnerer ikke noe, den skriver bare ut. Ofte vil vi lage funksjoner som returnerer noe

```python
def forrentning(startsum, rente, år):
    return startsum*(1+rente)**år

fremtidssaldo = forrentning(100000, 0.03, 20)
```

Tidligere har vi laget lister for kontoinformasjon. For eksempel:

```python
kunde_info = ["Trude Larsen", 23, 23433.45, False]
print(kunde_info[2])
```

Der det er implisitt at det førsste elementet er kundens navn, det andre elementet er kundens alder, det tredje saldoen og det fjerde hvorvidt kunden er en viktig kunde. Med dictionarys kan dette gjøres mer elegant:

Vi kan lage lister av dictionaries.

```python
kundeliste = []
kundeliste.append({"navn":"Trude Larsen", "alder":23, "saldo":23433.45, "viktig":False})
print(kundeliste[0]["saldo"])
kundeliste.append({"navn":"Maryam Mirzakhani", "alder":46, "saldo":12223333.41, "viktig":False})
kundeliste.append({"navn":"Sofya Kovalevskaya", "alder":79, "saldo":33423432.85, "viktig":False})
kundeliste.append({"navn":"Nariman Gulsvik", "alder":14, "saldo":433.75, "viktig":False})
kundeliste.append({"navn":"Gunnar Fidjestøl", "alder":53, "saldo":34453.47, "viktig":False})
print(kundeliste[3]["viktig"])
```

Banken har bestemt at kundenes rente også skal lagres i kundeinformasjonen. Til å begynne med settes alles rente til 0.02. Det kan vi gjøre ved å bruke en for løkke:

```python
for kunde in kundeliste:
    kunde["rente"]= 0.02
```

```python
for kunde in kundeliste:
    if(kunde["saldo"] >100000):
        kunde["viktig"] = True
#print(kundeliste)
```

I tillegg til lister og dictionaries, har python datatypene lister og tupler. Tupler ligner på lister, men en viktig forskjell er at elementene ikke kan endres. 

```python
my_tuple = (1,4) #Legg merke til at vi bruker vanlige parenteser
print(my_tuple[1])
```

```python
my_tuple[1] = 3
```

Dersom vi har data vi ikke vil skal endres, kan det være lurt å bruke tupler. Tupler er også raskere enn lister. Den fjerde og siste innebyggde datatypen i python er sett. Sett er ikke ordnet og har ikke mulighet for gjentatte elementer.  

```python
my_set = {1.0, 1.0, "hei", "hei"}
print(my_set)
```

```python
print(my_set.intersection({"hei", "nei"}))
```
