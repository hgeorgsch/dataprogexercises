# Matplotlib - Barplot

Jonas har en fruktgård og produserer epler og pærer.

Vi lager et program som viser hvor mange tonn epler og pærer som blir produsert på gården mellom 2012 og 2020

```python
import matplotlib.pyplot as plt
import numpy as np

n = 9 # Antall år
tid = []
start_year = 2012
for i in range(n):
    tid.append(start_year+i)

prod_epler = [22,54,37,35,23,35,19,30,31]
prod_pears = [66,67,89,45,47,57,34,22,19]
total_produksjon = []
for i in range(n):
    total_produksjon.append(prod_epler[i]+prod_pears[i])

plt.plot(tid,prod_epler)
plt.plot(tid,prod_pears)
plt.plot(tid,total_produksjon)
plt.show()
```
