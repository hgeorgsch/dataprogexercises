## f-strenger
Ved å bruke f-strenger kan man sette uttrykk rett inn i en streng:

```python
minstreng = f'De første partallene er {a[1:]}'
print(minstreng)
```

```python
def f(x):
    return 4*x -1
nystreng = f'Tallet er {f(2)+3}'
print(nystreng)
```

Dersom vi kombinerer det vi har lært om f-strenger og slicing kan vi behandle datoer på en fin måte: 

```python

dato = "25.12.1642"
year = int(dato[-4:])
print(year)
dag = int(dato[:2])
print(dag)
month = int(dato[3:5])
print(month)
month_list = ["ugyldig", "Januar", "Februar", "Mars", 
              "April", "Mai", "Juni", "Juli", "August", "September", "Oktober", "November", "Desember"]
print(f"Sir Isaac Newton ble født den {dag}. {month_list[month]} herrens år {year}")
```

Vi er interesserte i å lagne en funksjon som tar inn en fødselsdadato og personens navn, og skriver ut hvor mange år etter utbruddet (i verden) av andre verdenskrig personen ble født.

```python
def år_etter_krigen(dato,navn):
    år = int(dato[6:])-1939
    print(f"{navn} ble født {år} etter krigen")
brandt_navn = "Julian Brandt"
brandt_dato = "02.05.1996"
år_etter_krigen(brandt_dato, brandt_navn)
```

## Listekomprehensjon
En effektiv måte å lage en liste på er å bruke listkomprehensjon. Her er en fin video om temaet :https://www.youtube.com/watch?v=AhSvKGTh28Q

```python
a = [i**2 for i in range(40)]
b = [i for i in a if i != 9 ]
print(a)
print(b)
```

La oss se på et eksempel med tekststrenger:



```python
kundeliste = ["Kari Knudsen", "Rosa Luxenburg", "Olga Romanchuk", "Indira Ghandi", "Karl Popper", "Victor Boniface"]
## Uten listekomprehensjon:
min_liste_uten_komprehension = []
for kunde in kundeliste:
    if (kunde != "Rosa Luxenburg"):
        min_liste_uten_komprehension.append(kunde)
        
print(min_liste_uten_komprehension)
#Med komprehension
med = [navn for navn in kundeliste if navn != "Rosa Luxenburg"]
med
```
