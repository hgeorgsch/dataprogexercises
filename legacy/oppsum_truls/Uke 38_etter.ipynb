# Tidsrekker
Til nå har vi jobbet med folketallsmodeller og sparing i øvingene. I den første mappen, kommer dere til å få en ny tidrekke. Først ser vi på et enkelt eksempel. I forelesningene har vi sett på det enkle eksemplelet at et engangsbeløp settes inn på en konto. Formelen for beløpet på konto er da:
$$P(t) = P_0(1+r)^t$$
Der $P_0$ er startbeløpet, $r$ er renten og $t$ er antall år. 

```python
def forrentning(r, startsum, tid):#forrentning gir hvor mye startsum vokser til med r prosenst rente i tid år.
    return startsum*(1+r)**tid

def forrentningsliste(r,startsum,sluttid):#forrentningsliste gir en liste over hvor mye startsum vokser hvert år til slutttid 
    saldoliste = []
    for tidspunkt in range(sluttid):
        saldoliste.append(forrentning(r,startsum,tidspunkt))
        
    return saldoliste

```

```python
print(forrentningsliste(0.05, 1000, 10))
```

Vi er interesserte i å undersøke tallene vi får fra tidsrekkene våre nærmere.


```python
def myAverage(my_values):
    my_sum = 0
    for value in my_values:
        my_sum += value
    return my_sum/(len(my_values))
print(myAverage([1,2,1.5,10,-100]))
```

Vi ønsker å lage en funksjon som lager en liste over beløpet, helt til man når sin målsum:

```python
def savingsList(rente, startsum, målsum):
    my_money = []
    my_money.append(startsum)
    penger_nå = startsum
    while penger_nå < målsum:
        penger_nå = penger_nå*(1+rente) 
        my_money.append(round(penger_nå))  
    return my_money
print(savingsList(0.01, 1000, 1500))
```

Det er viktig at vi kan plotte dataene våre:


Vi kan finne antallet år det tar å nå målsummen ved å bruke len()

```python
print(len(savingsList(0.01,1000, 1000000)))
```

```python
import matplotlib.pyplot as plt
import numpy as np
my_saving = savingsList(0.35, 1000, 20000)
år = np.linspace(0,len(my_saving), len(my_saving))
plt.plot(år, my_saving)
```
