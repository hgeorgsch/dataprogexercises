# Oppgave 8

----

Dersom vi sparer et fast terminbeløp $P$ som vi betaler inn på konto $n$ ganger i året, vi har en p.a. rente på $r$ prosent og vi sparer i $t$ år, vil sluttbeløpet på sparekontoen være gitt ved:
$$
F = P\cdot \left(\frac{\left(1+\frac{r}{n}\right)^{nt}-1}{\frac{r}{n}}\right)
$$

Denne oppgavene er tredelt
* Først skal dere anta at man setter inn 1000 kr 12 ganger i året og renten er 3 prosent. Bruk pyplot til å lage et plot over beløpet på kontoen som en funksjon av tiden frem til det har gått 10 år.
* Anta så at dere etter inn etter inn 1000 kr 12 ganger i året Anta at renten er mellom 0 og 10 prosent. Plot beløpet etter 10 år som en funskjon av renten. 
* Anta så at dere setter inn penger 12 ganger i året i 10 år. Anta at renten er 3 prosent. Plot sluttbeløpet som en funksjon av det månetlige innbetalingsbeløpet. Anta at det månetlig innbetalingsbeløpet er mellom 500 kr og 3000 kr. 
* Finn ut hvor mange år man må spare for å få like mye på kontoen som  dersom man setter inn 1000 kr 12 ganger i året og renten er 4 prosent og sparer i ti år, dersom man setter inn 1000 kr 12 ganger i året og renten er 3 prosent. 

*(Merk at $r$ i formelen ikke er i prosent, 2% = 0.02)*
