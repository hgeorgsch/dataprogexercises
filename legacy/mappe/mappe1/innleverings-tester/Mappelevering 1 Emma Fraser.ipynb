# Oppgavetekst til oppgave 2: 

 Dersom vi sparer et fast terminbeløp $P$ som vi betaler inn på konto $n$ ganger i året, vi har en p.a. rente på $r$ prosent og vi sparer i $t$ år, vil sluttbeløpet på sparekontoen være gitt ved:
$$
F = P\cdot \left(\frac{\left(1+\frac{r}{n}\right)^{nt}-1}{\frac{r}{n}}\right)
$$

Lag minst fire numpy dictionaries. Hver dictionary er for en kunde. Dictionariene skal inneholde:
* Kontoinnehavernes navn. 
* Kontoinneavernes terminbeløp.
* Sluttbeløp på sparekontoen.
* Antall år spart
*(Merk at $r$ i formelen ikke er i prosent, 2% = 0.02)*

* Lag en if setning

# Oppgave 3: Lånekalkulator

# Oppgavetekst til oppgave 3:

Vi tar opp et annuiteteslån på 600 000kr til en årlig rente på 4%. Lånet skal tilbakebetales i 12 årlige og like store terminbeløp, det første om ett år. Hva blir terminbeløpet og den totale kostnaden for lånet?

Denne oppgaven skal inneholde:
*  Utregning av årlig terminbeløp
*  Liste av kostnader, rente og avdrag hvert år
*  Utregning av totalkostnad for hele lånet

```python
#Oppgave 3 lånekalkulator 

# Formel terminbeløp

D = (K*r*(1+r)**n)/((1+r)**n-1)

#Informasjon

K = 600000
r = 0.04
n = 12

#Utregning av årlig terminbeløp

print("Det årlige terminbeløpet er:")
print(D)
print("\n")

#Totalekostnad for lånet:
print("Totalkostnaden for lånet er:")
totalkostnad = D * n
print(totalkostnad)

print("\n")
print("Rentekostnaden for lånet er:")
rentekostnad = totalkostnad - K
print(rentekostnad)

print("\n")

# Informasjon
K = 600000
r = 0.04
n = 12

# Utregning av årlig terminbeløp
D = (K * r * (1 + r)**n) / ((1 + r)**n - 1)

# Initiell restgjeld
restgjeld = K

# Nedbetalingsplan
nedbetalingsplan = []

print(f"{'År':<5}{'Terminbeløp':<15}{'Rentekostnad':<15}{'Avdrag':<15}{'Restgjeld':<15}")
print("-" * 70)

for år in range(1, n + 1):
    rentekostnad = restgjeld * r
    avdrag = D - rentekostnad
    restgjeld -= avdrag
    
    nedbetalingsplan.append({
        'År': år,
        'Terminbeløp': D,
        'Rentekostnad': rentekostnad,
        'Avdrag': avdrag,
        'Restgjeld': max(0, restgjeld)  # Unngå negative restgjeldsverdier
    })

    print(f"{år:<5}{D:<15.2f}{rentekostnad:<15.2f}{avdrag:<15.2f}{max(0, restgjeld):<15.2f}")
```
