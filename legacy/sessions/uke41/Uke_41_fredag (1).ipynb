# Pandas
Pandas er et bibliotek som brukes veldig mye til å behandle data. Datframes har kolonner og rader. La oss se på det gode gamle eksempelet med kontolister. I Pandas bruker vi noe som heter dataframes. 

```python
import pandas as pd
kontoinnehaver_liste = pd.DataFrame(["Uwe Seeler", "Jürgen Kohler", "Andres Iniesta", "Heidi Tjugum", "Ivan Myshkin", "Nastasha Filipovna"])
print(kontoinnehaver_liste)
```

Som i dictionaries kan vi gi nøkkelord til kolonner. 


```python
kontoinnehaver_liste = pd.DataFrame({"Navn" :["Uwe Seeler", "Jürgen Kohler", "Andres Iniesta", "Heidi Tjugum", "Ivan Myshkin", "Nastasha Filipovna"]})
```

```python
print(kontoinnehaver_liste)
```

### Legge til kolonne med insert
Det finnes flere måter man kan legge til en kolonne i et dataframe. En måte er ved å bruke insert. Argumentene til insert funksjonen er (plassering av kolonnen, Navnet til kolonnen, verdiene i kolonnen)

```python
kontoinnehaver_liste.insert(1,"Viktig", [False, True, True, False, False, True])
print(kontoinnehaver_liste)    
```

For å velge en rad kan vi bruke iloc. 

```python
print(kontoinnehaver_liste.loc[2])
```

For å velge kolonne kan vi slå opp på nøkkelordet

```python
print(kontoinnehaver_liste["Navn"])
```

Vi ønsker så å legge til alder til de ulike kundene. Vi kan gjøre dette med insert, eller vi kan legge kolonnen til på slutten av dataframet ved å bruke kolonnenavnet:


```python
kontoinnehaver_liste["Alder"] = [89,67,39, 48, 23, 21]
print(kontoinnehaver_liste)
```

```python
kontoinnehaver_liste.insert(2, "Saldo",[2323,23233434,35355435,45656,56767,345634634656646])# [23233.20, 100000222, 2222222222222, 2342, 23, 23423443245])
print(kontoinnehaver_liste)
```

Vi kan også bruke loc til å legge til en kolonne. 

```python

kontoinnehaver_liste.loc[:,"rente"] = [0.12, 0.12,0.34, 0.34, 0.01,0.35]
```

```python
print(kontoinnehaver_liste)
```

Vi kan også legge til en kolonne der vi bruker den forrige

```python
kontoinnehaver_liste["Saldo neste år"] =(1+ kontoinnehaver_liste["rente"])*kontoinnehaver_liste["Saldo"] 
```

```python
print(kontoinnehaver_liste)
```

Dersom vi vil legge til en rad kan vi bruke loc. 

```python
kontoinnehaver_liste.loc[6] = ["Gregor Samsa", False, 2344,25,  0.00, 2344]
kontoinnehaver_liste
```

### Manipulering av lister
Vi ønsker å la alle kontoinnehaverne bli 10 år yngre:

```python
kontoinnehaver_liste["Alder"] = kontoinnehaver_liste["Alder"] - 10
```

```python
kontoinnehaver_liste
```

Generelt kan vi bruke loc til å få tligang til et element i dataframet vårt, og til å endre elementet, dersom vi er interesserte i det. Det er viktig at man angir radnummeret først, og kolonnen på den andre posisjonen. 

```python
kontoinnehaver_liste.loc[ 1, "Saldo"] = 2
kontoinnehaver_liste.loc[ 2, "Alder"] 
```

La os si at vi vil lage en prognose for hva de ulike gruppene kommer til å tjene om 10 år. Det kan vi gjøre på samme lignende måte som vi har manipulert dataframes før. 

```python
lønn["prognose 2032"] = lønn[2022]*(1+lønn['Endring i prosent']/100)**10
lønn["prognose 2032"] = lønn["prognose 2032"].round(2)
(lønn)
```

Nå vil vi regne ut gjennomsnittlig inntekt i 2022 og i 2032:

```python
mean_2022 = round(lønn[2022].mean(),2)
mean_2032 = round(lønn["prognose 2032"].mean(),2)
print(f"I Norge var gjennomsnittslønnen i 2022 {mean_2022} NOK, men gjennomsnittslønnen etter vår prognose vil være {mean_2032} NOK i 2032") 
```
