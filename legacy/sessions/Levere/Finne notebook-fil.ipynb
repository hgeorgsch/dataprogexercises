# Finne riktige jupyter-notebook-filer (*.ipynb)

* For å finne tak i jupyter-notebook filene deres kan dere:
  - Finne fram til riktig filplassering i windows-explorer/File explorer eller Finder på mac
  - «Laste ned» notebooken, da lagres den i nedlastningsmappen til nettleseren du bruker

## Finne riktig filplassering

* «Default» plassering når du åpner jupyter-notebook med anaconda er:
  - `C:\Users\<brukernavn>` på windows - brukernavn er brukeren på PC'n, ikke ntnu-brukernavnet ditt
  -  `/users/<brukernavn>/` på macos - brukernavn er brukeren på Macen, ikke ntnu-brukernavnet ditt

### Med `OS`
* Dersom du ikke finner filene under «standard» plassering, kan du kjøre koden under i en notebook du forsøker å finne
* Her printes stien som notebooken ligger i ut
