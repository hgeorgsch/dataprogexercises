# Gjennom statistikkbanken 1

* Finn frem til dataene du vil ha på vanlig måte
* I stedet for å lagre data kan man trykke lagre spørring
* Vi velger hvilke tidspunkt vi vil ha samt format (json_stat2)
* Og får en lenke vi kan bruke til å hente dataen med

```python
import requests
import pandas as pd
from pyjstat import pyjstat
#!pip install pyjstat
#Kommenter ut linjen over for å installere pyjstat

lagret_url = "https://www.ssb.no/statbank/sq/10089394"

response = requests.get(lagret_url)

# Gjør om til pandas dataframe:
dataset = pyjstat.Dataset.read(response.text)
df_id = dataset.write('dataframe', naming='id')
df = dataset.write('dataframe')

```

```python
#Med json_stat trenger vi gjerne å bruke pivot() funksjonen
df_ny = df.pivot(columns="år", values="value", index=["kjønn", "alder"])

#Unstack flytter index opp til kolonnene (det blir flere kolonnenivåer)
df2 = df_ny.unstack(0)

#Vi kan også trikse inn kjønn som egen kolonne
df3 = df_ny.reset_index().set_index("alder")
```

# Gjennom statistikkbanken 2
* Vi kan bla nederst på siden til tabellen vi har laget i statistikkbanken
* og trykke "API for denne tabellen"
* Vi kopierer url og laster ned spørring som JSON fil

```python
import json

#url = "https://data.ssb.no/api/v0/no/table/06129/"


with open("ssbapi_table_06129.json") as file:
    query = json.load(file)

tabell = query["tableIdForQuery"]
post_url = query["postUrl"]
query_post = query["queryObj"]

response = requests.post(post_url, json=query_post)
data = response.json()
```
