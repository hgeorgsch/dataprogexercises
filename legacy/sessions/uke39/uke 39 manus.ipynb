# Matplotlib - Barplot

Jonas har en fruktgård og produserer epler og pærer.

Vi lager et program som viser hvor mange tonn epler og pærer som blir produsert på gården mellom 2012 og 2020

```python
import matplotlib.pyplot as plt
import numpy as np

n = 8 # Antall år
tid = []
start_year = 2012
for i in range(n+1):
    tid.append(start_year+i)

prod_epler = [22,54,37,35,23,35,19,30,31]
prod_pears = [66,67,89,45,47,57,34,22,19]
total_produksjon = []
for i in range(n+1):
    total_produksjon.append(prod_epler[i]+prod_pears[i])

plt.plot(tid,prod_epler, label="epler")
plt.plot(tid,prod_pears, label="pears")
plt.plot(tid,total_produksjon, label="Total")
plt.xlabel("År")
plt.ylabel("Produksjon [tonn]")
plt.legend()
plt.title("Jonas' fruktproduksjon")
plt.show()
#barplot ville nok vært bedre...
```

```python
barx_pos = np.arange(len(prod_epler))
start_year = 2012
tid = []
for i in range(n+1):
    tid.append(str(start_year+i))
    
#align edge putter venstre kant på 'x'    
plt.bar(barx_pos, prod_epler, width=0.25, align="edge", label="epler")

#align edge putter høyre kant på 'x' når width er negativ
plt.bar(barx_pos, prod_pears, width=-0.25, align="edge", label="pears")
plt.xlabel("År")
plt.ylabel("Produksjon [tonn]")
plt.xticks(barx_pos, tid) # Setter riktig merkelapp på x-aksen
plt.legend()
plt.title("Jonas' fruktproduksjon")
plt.show()
```

```python


bar_bredde = 0.7/3
barx1 = np.arange(len(prod_epler))
barx2 = []
barx3 = []
for x1 in barx1:
    barx2.append(x1-bar_bredde/2)
    barx3.append(x1+bar_bredde/2)


   
plt.bar(barx1, prod_epler, width=bar_bredde, align="center", label="epler")
plt.bar(barx2, prod_pears, width=-bar_bredde, align="edge", label="pears")
plt.bar(barx3, total_produksjon, width=bar_bredde, align="edge", label="Total")
plt.xlabel("År")
plt.ylabel("Produksjon [tonn]")
plt.xticks(barx_pos, tid) # Setter riktig merkelapp på x-aksen
plt.legend()
plt.title("Jonas' fruktproduksjon")
plt.show()
```

```python
barx_pos = np.arange(len(prod_epler))
start_year = 2012
tid = []
for i in range(n+1):
    tid.append(str(start_year+i))
    
#align edge putter venstre kant på 'x'    
plt.bar(barx_pos, prod_epler,label="epler")

#align edge putter høyre kant på 'x' når width er negativ
plt.bar(barx_pos, prod_pears,label="pears", bottom=prod_epler)
plt.xlabel("År")
plt.ylabel("Produksjon [tonn]")
plt.xticks(barx_pos, tid) # Setter riktig merkelapp på x-aksen
plt.legend()
plt.title("Jonas' fruktproduksjon")
plt.show()
```
