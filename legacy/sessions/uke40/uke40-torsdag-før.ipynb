## Hypotesetesting:

* Hypotese: Dersom du er 'fullblods' har du større sjanse å overleve/krepere enn om du er gomp eller halvblods

Vi tester:
$$
\begin{align}
H_0: \mu &= \mu_0 \quad \text{mot} & H_1: \mu > \mu_0
\end{align}
$$
Vi sjekker om en trollmann er i live, $0$ for død, og $1$ for levende, eventuelt omvendt om vi vil måle hvor "død"
trollmannen er.

$H_0$, nullhypotesen, sier at forventningsverdien til $\mu$ når vi sjekker levende/døde fullblods, er den samme som forventningsverdien $\mu_0$ for alle trollmenn.

Vi påstår at $H_1$, at fullblods har større sjanse til å overleve/dø, dersom for den «målte» $\hat\mu$:
$$
\hat\mu > k = \mu_0+z_{\alpha}\frac{\sigma}{\sqrt{n}}
$$

Her er $z_{\alpha}$ er antall standardvik unna forventningsverdi som gir $\alpha$ igjen i normalfordelingen,
$\sigma$ er standardavvik og $n$ er antall fullblodstrollmenn vi sjekker
