## Dictionaries

Dictionaries var en måte å pakke samme data på $\implies$ datastruktur

```python

person = {"navn": "Jonas J. Harang",
         "alder": 36,
         "stilling": "Universitetslektor",
         "Institutt": "IIR",
         "overarbeid": True,
         "underkompensert": False}
print(person)
```

* Feltene med "navn på data" kalles keys ("navn", "alder", "stilling"...)
* Datafeltene kalles "values", til hver "key" hører det til 1 "value"
* (Men den kan være `None` $\implies$ ingenting)

* Man får tak i data ("value") som tilhører et kategori ("key") ved å bruke `dictionary["key"]`
* Man kan også legge til nye datapunkt på samme måte
* Man kan også legge til felt eller slå sammen dictionaries med den innebygde metoden `update`

```python
print("Henter ut navn:",person["navn"])
#Legger til et nytt element:
person["mellomnavn"] = "Julius"

#Oppdaterer et element
person["navn"] = "Jonas Harang"

#Legger til OG oppdaterer (slår sammen dictionaries)
person.update( {"fag": ["matematikk", "fysikk", "programmering"], "navn": "Harang, Jonas"})

print("mellomnavn", person["mellomnavn"])
print("Navn:", person["navn"], "\n", "Fag:", person["fag"])
```

* Det er ikke uvanlig å ha dictionaries inne i dictionaries

```python


person.update({
"stilling": {
             "navn": "Universitetslektor",
             "ansatt": "14.12.20",
             "stillingsprosent": "100%"
            }
})
print(person)
```

## Hypotesetesting:

* Hypotese: Dersom du er 'fullblods' har du større sjanse å overleve/krepere enn om du er gomp eller halvblods

Vi tester:
$$
\begin{align}
H_0: \mu &= \mu_0 \quad \text{mot} & H_1: \mu > \mu_0
\end{align}
$$
Vi sjekker om en trollmann er i live, $0$ for død, og $1$ for levende, eventuelt omvendt om vi vil måle hvor "død"
trollmannen er.

$H_0$, nullhypotesen, sier at forventningsverdien til $\mu$ når vi sjekker levende/døde fullblods, er den samme som forventningsverdien $\mu_0$ for alle trollmenn.

Vi påstår at $H_1$, at fullblods har større sjanse til å overleve/dø, dersom for den «målte» $\hat\mu$:
$$
\hat\mu > k = \mu_0+z_{\alpha}\frac{\sigma}{\sqrt{n}}
$$

Her er $z_{\alpha}$ er antall standardvik unna forventningsverdi som gir $\alpha$ igjen i normalfordelingen,
$\sigma$ er standardavvik og $n$ er antall fullblodstrollmenn vi sjekker

### **NB! Tvilsom statistikk**

```python
import scipy.stats as st
import numpy as np

alpha = 0.98
z_alpha = st.norm.ppf(alpha)


pp.pprint(HPdata[0])

        
total_lever = []
for person in HPdata:
    if person["alive"]:
        total_lever.append(0.0)
    else:
        total_lever.append(1.0)

lever_fullblods = []
for person in HPdata:
    if person["ancestry"] == "pure-blood":
        if person["alive"]:
            lever_fullblods.append(0.0)
        else:
            lever_fullblods.append(1.0)
            
            
mu_0 = np.mean(total_lever)
mu = np.mean(lever_fullblods)

print("Forventningsverdi alle", mu_0)
print("Forventningsverdi, fullblods", mu)
print("Tot standardavvik", np.std(total_lever))
print("Fullblods standardavvik", np.std(lever_fullblods))

n = len(lever_fullblods)
std_fullblods = np.std(lever_fullblods)
print("Standardavvik", std_fullblods)
k = mu_0+z_alpha*std_fullblods/np.sqrt(n)
print("k", k)

if mu > k:
    print("Det ser ut som fullblods-trollmenn lever kortere enn andre")
    print("Det kan vi si med", int(alpha*100), "prosent sikkerhet")
else:
    print("Vi kan ikke konkludere med at fullblodstrollmenn levere kortere, dersom vi krever å være", int(alpha*100), "prosent sikker på konklusjonen")
```
