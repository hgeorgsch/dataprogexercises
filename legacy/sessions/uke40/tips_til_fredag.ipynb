```python
import matplotlib.pyplot as plt
import numpy as np
import pickle
import pprint

#Lager et "pprint" objekt som printer med innrykk på 4:
pp = pprint.PrettyPrinter(indent=4)
  

with open("HPcharacters.pickle", 'rb') as file:
    HPdata = pickle.load(file)

pp.pprint(HPdata)




```

```python
# listekomprehensjon

karakterliste = [karakter["name"] for karakter in HPdata]
hogwart_teachers = [karakter["name"] for karakter in HPdata if karakter["hogwartsStaff"]]
in_memoriam = [karakter["name"] for karakter in HPdata if karakter["hogwartsStaff"] and not karakter["alive"]]
print(in_memoriam)
```

```python
#Dictionary/Set komprehensjon

hus_set = set({karakter["house"] for karakter in HPdata})
hus_set2 = set([karakter["house"] for karakter in HPdata])
hus_set_fix = set([karakter["house"] for karakter in HPdata if len(karakter["house"])>0])
print(hus_set)
print(hus_set_fix)

#Dictionary med medlem av forskjellige hus
HP_by_house = {k : [] for k in hus_set_fix}
for karakter in HPdata:
    hus = karakter["house"]
    if hus != '':
        HP_by_house[hus].append(karakter["name"])
        
#Skikkelig lang en
HP_by_house = { k: [karakter["name"] for karakter in HPdata if karakter["house"]==k] 
               for k in hus_set_fix
              }

pp.pprint(HP_by_house)

```
