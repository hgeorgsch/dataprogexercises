```python
class Agent:
    def settPosisjon(self,x,strand):
        self.pos = x
        self.strand = strand
    def fåPosisjon(self):
        return self.pos
    def handle(self):
        print( "ikke implementert" )

class Badegjest(Agent):
    def handle(self):
        v = self.strand.finnIskrem(self.pos)
        v.betal( random.randint( 25, 150 ) )

class Iskremvogn(Agent):
    def __init__(self):
        self.omsetjing = 0
        self.historie = [ 0 ]
    def betal(self, b):
        self.omsetjing += b
    def handle(self):
        if self.omsetjing < self.historie[-1]:
            self.pos = self.strand.move( p )
        self.historie.append( self.omsetjing )
        self.omsetjing = 0
class Strand:
    def move(self,p):
        return p
    def __init__(self,lengde=1000):
        self.lengde = lengde
        self.badegjester = []
        self.iskremvogner = []
    def finnIskrem( self, pos ):
        bestevogn = None
        bestepos = 2**32
        for v in self.iskremvogner:
            p = v.fåPosisjon()
            if abs( p - pos ) < bestepos:
                bestepos = abs( p - pos )
                bestevogn = v
        return bestevogn
    def leggtilBadegjest(self,bg):
        x = random.randint(0,self.lengde)
        bg.settPosisjon( x, self )
        self.badegjester.append( bg )
    def leggtilIskremvogn(self,iv):
        x = random.randint(0,self.lengde)
        iv.settPosisjon( x, self )
        self.iskremvogner.append( iv )
    
class Simulator:
    def __init__(self,ngjest,nvogn,lengde=1000):
        self.badegjester = [ Badegjest() for _ in range(ngjest) ]
        self.iskremvogner = [ Iskremvogn() for _ in range(nvogn) ]
        self.strand = Strand(lengde=lengde)
        for g in self.badegjester: self.strand.leggtilBadegjest(g)
        for v in self.iskremvogner: self.strand.leggtilIskremvogn(v)

    def køyr(self,n=100):
        for t in range(n):
            for g in self.badegjester: g.handle()
            for g in self.iskremvogner: g.handle()
            
```
