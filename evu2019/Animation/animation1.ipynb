```python
fig = plt.figure()
ax = plt.axes(xlim=(-2, 2), ylim=(-2, 2))
circle = plt.Circle((0,0), 0.25, color='blue')
ax.add_artist(circle)
```

```python
def animate(i):
    (x,y) = circle.get_center()
    circle.set_center((x+0.05,y+0.05))
    fig.canvas.draw()
    return circle,

anim = animation.FuncAnimation(fig, animate, frames=200, interval=20, blit=True)
```
