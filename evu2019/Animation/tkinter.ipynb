Der er mange måtar å bruka grafikk på i python og jupyter.

Me skal bruka biblioteket ``tkinter``, so me startar med å importera dette.


```python
import tkinter as tk
```

Me har importert ``tkinter`` under namnet ``tk``.
For å bruka funksjonar i biblioteket må me bruka prefiksen ``tk.``.
Dette lét oss halda funksjonar frå ulike bibliotek frå kvarandre.

## Nokre statiske døme

### Døme 1

Lat oss ta eit døme:

```python
vindauga = tk.Tk()

lerret = tk.Canvas(vindauga, width=800, height=400)
lerret.pack()

y = 200
lerret.create_line(100, y, 700, y, fill="#476042")

vindauga.mainloop()
```

Dette dømet er forskjellig frå det me har sett tidlegare.
Dette er eit heilt program som ikkje let seg køyra ei line åt gongen.

Den fyrste lina opnar eit vindauga, som køyrer heilt til brukaren lukkar det.  Den andre linen lagar eit lerret, til å teikna på, inni vindauga.

På lerretet teiknar me ein strek, med ``create_line``-funksjonen.

Ingenting skjer før vindauga vert køyrd, med ``mainloop``-funksjonen.

### Døme 2

Før du kan køyra døme 2, må du lukka vindauga frå døme 1.

```python
gui = tk.Tk()
gui.geometry("400x400")
gui.title("First title")
canvas = tk.Canvas(gui, width=300,height=300,bg='black')
canvas.pack()
ov1 = canvas.create_oval(5,5,60,60, fill='pink')
gui.mainloop()
```

Dette dømet viser eit par andre funksjonar. Me kan setja storleik og tittel på vindauga, og me kan teikna ein oval i staden for ei line.
Elles er prinsippa dei same.

## Animasjon

### Døme 3

```python
import time

gui = tk.Tk()
gui.geometry("800x800")
c = tk.Canvas(gui ,width=800 ,height=800)
c.pack()
oval = c.create_oval(5,5,60,60,fill='pink')
a = 5
b = 5
for x in range(0 ,100):
  c.move(oval,a,b)
  gui.update()
  time.sleep(.01)
gui.title("Second title")
gui.mainloop()
```
