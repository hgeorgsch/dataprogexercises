## Slumptalsgeneratoren

I denne øvinga skal me bruka to nye byggeklossar.

```python
rnd.rand()
```

```python
rnd.randn()
```

I python-jargon er `rnd.rand()` og `rnd.randn()` funksjonar, men matematisk sett er dei det ikkje. Dei returnerer nemleg ikkje det same svaret kvar gong. For å få eit inntrykk, kan me sjå kva som skjer når me køyrer dei mange gongar.

```python
for i in range(10): print(i, rnd.rand())
```

```python
for i in range(10): print(i, rnd.randn())
```

Båe funksjonane er slumptalsgeneratorar. Dei gjev tal som *skal sjå tilfeldige ut*. Kva slags sannsynsordeling trur du dei to genoratorane simulerer?

## Funksjon for terningkast

Me hugsar at me kasta 25 terningar slik:

```python
x = np.floor(rnd.rand(25)*6+1)
x = x.astype(int)
```

Det er nyttig med ein funksjon for dette, slik at me kan kasta terning mange gongar.

```python
def dice(n): return np.floor(rnd.rand(n)*6+1).astype(int)
```

Merk at `dice()` returnerer ein matrise:

```python
dice(10)
```

Det er enkelt å rekna ut gjennomsnitt og standardavvik på slike matrisar:

```python
x = dice(10)
print(x)
print(np.mean(x))
print(np.std(x))
```

Det er i seg sjølv ein interessant øving å observera mange gjennomsnitt, t.d. slik:

```python
print( [ np.mean(dice(10)) for i in range(20)] )
```
