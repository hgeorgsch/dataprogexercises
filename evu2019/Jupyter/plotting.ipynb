## Funksjonsdefinisjonen

Lat oss sjå på tredjegradsfunksjonen $$f(x) = 3x^3 + x^2 -3.$$
Me kan definera ein python-funksjon som evaluerer funksjonen $f(x)$, som fylgjer.


```python
def f(x): return 3*x**3 + x**2 - 3
```

Legg merke til bruken av aritmetiske operatorar, og ** for potens.

Me kan testa denne funksjonen i python:

```python
f(0.5)
```

Merk at me ikkje utan vidare kan få fleire verdiar frå same blokk.

```python
f(-1)
f(+1)
```

Jupyter viser berre den siste verdien i blokken.
Dersom me skal testa fleire verdiar og skriva ut meir informasjon, kan me bruka ``print``-funksjonen:

```python
print ("f(-1) =", f(-1))
print ("f( 0) =", f(0))
print ("f(+1) =", f(1))
```

Me har sett tre ulike typar kodeliner over:
1. Definisjon ``def f(x)``: funksjonen $f(x)$ vert definert, men ingenting vert skrive ut.
2. Evaluering ``f(0.5)`` på eiga line: funksjonen vert evaluert og me ser resultatet.
3. Kommando ``print``: python får ein kommando, som i dette tilfellet skriv ut argumenta.
Merk at dersom kodeblokken inneheld fleire evalueringar, vert berre den siste vist:

```python
f(1)
f(0)
```

Me bruker ``print`` får å få kontroll over visinga.

Ein kommando og ei evaluering er eigentleg det same i python.  Skilnaden er at ``f(x)`` gjev ein returverdi som python kan bruka vidare, men ``print(x)`` skriv på skjermen utan å gje ein returverdi. Det siste er ein sokalla bieffekt
(*side effect*).
Det er mogleg for ein funksjon å gje både returverdi og bieffekt, men mange reknar det som dårleg praksis.

## Variablar og tilordning

Ein av dei mest fundamentale setningane i programmering er tilordning av variablar:

```python
y = f(-2)
```

OK.  Noko er kanskje skjedd her, men me veit ikkje kva.  Lat oss sjå.

```python
y
```

Det er viktig å merka seg at notasjonen x = y i python ikkje er eit utsagn som kan vera sant eller usant (samanlikning).  Det er ein imperativ som seier at noko skal skje (tilordning).
Etter tilordninga har $x$ (variabelen på venstre side) fått ein ny verdi.

I Jupyter skal ein òg leggja merkje til at ein kan køyra blokkane i feil rekkjefylgje. Variablar vert tilordna når blokken vert køyrd.  Difor er det ikkje sikkert at variablane har den verdien det ser ut som om dei har når ein les
dokumentet som det er skrive.  Lat oss sjå på eit døme.

Lat oss fyrst tilordna:

```python
a = 1
```

So kan me bruka variabelen, t.d.

```python
f(a)
```

Her burde $f(a)=1$.  Lat oss no endra verdien på $a$.

```python
a = 2
f(a)
```

No bør $f(a) = 25$, men legg merke til at den forrige utrekninga ikkje er endra no.  Me kan derimot gå tilbake og rekna den forrige blokken på nytt.
Dersom me gjer det (utan å køyra tilordninga på nytt), får me $f(a)$ ogso der.

For å vera sikker på at alt er rekna ut i riktig rekkjefylgje kan ein køyra heile dokumentet i rekkjefylgje.
*Spol fram*-knappen (dobbel pilhovud til høgre) i Jupyter gjer dette.

**Tips:**
Det er ein god vane, når ein lurer litt på korleis ein programkonstruksjon verkar, å prøva seg fram. 
Skriv nokre enkle døme og prøv å køyra dei.  
Ein lærer meir av døme enn av reglar.

**Merknad**
Tilordinga ``a=1`` er òg ein kommando.  Me kommanderer python til å endra verdien på ``a``.
Som me ser over gjev ikkje tilordninga nokon returverdi.

# Listekomprehensjon

```python
x = range(0,5)
print(x)
```

Funksjonen `range` gjev ein iterator. Me kan tenkja på det som ei liste med tal, frå 0 til 4 i dette tilfellet, men python reknar ikkje ut tala før det trengst, og `print` viser difor ikkje kva tal me får, berre korleis fylgja er definert.

Me kan laga ei liste i staden, vha. listekomprehensjon.

```python
x = [ i for i in range(0,5)]
print(x)
```

Listekomprehensjon er eit kraftig verkty og enkel notasjon for å rekna ut talfylgjer, ta t.d.

```python
y = [ 2*i+1 for i in x]
print(y)
```

Legg merke til parallellen mellom listekomprehensjon her, og mengdekomprehensjon i matematikken.
Matematisk ville eg skriva definisjonen over som
$$ y = [ 2i+1 | i \in x ] $$
Dette er litt uvanleg notasjon, sidan me sjelden arbeider med ordna lister i matematikken.
Hadde $x$ og $y$ vore uordna mengder ville me skriva
$$ y = \{ 2i+1 | i \in x \} $$

## Funksjonsplott

Lat oss sjå korleis me kan plotta funksjons $f(x)$ med plot-funksjonen som me har prøvd.
Fyrst må me definera $x$- og $y$-verdiar til plottet.
+ som $x$-verdiar kan me bruka (heil)tala i intervallet $[-10,10)$.
+ $y$-verdian må me rekna ut for kvar $x$-verdi.

```python
x = range(-10,10)
y = [ f(i) for i in x ]
print( "x:", x )
print ( "x:", [i for i in x ] )
print( "y:", y )
```

Då er me klare til å plotta.

```python
plt.plot(x,y,'+:')
```

Det er sjølvsagt nyttig å kunna plotta med tettare $x$-verdiar enn berre heiltalsverdiar.
Korleis kan me gjera det enkelt?
Me kan i alle fall kombinera range() og *list comprehension*.

```python
x = [ i*0.5 for i in range(-20,20)]
y = [f(i) for i in x]
plt.plot(x,y,'+')
```

```python
x = [ i*0.2 for i in range(-50,50)]
y = [f(i) for i in x]
plt.plot(x,y,'-')
plt.axes()
```

### Oppgåve

Tilpass $x$-verdiane og lag eit plot som tydleg viser tredjegradsformen med tre nullpunkt, topp- og botnpunkt.

## Visuell tilpassing


### Aksekors

Mange er vande med å sjå plottet i forhold til linene $x=0$ og $y=0$, heller enn innramma i ein boks som pyplot gjer i utgangspunktet.
Her er pyplot svært fleksibelt.  Alt er mogleg, til gjengjeld krev det nokre liner å seia akkurat kva me ynskjer.
Lat oss fyrst flytta $x$-aksen.

```python
plt.plot(x,y)
ax = plt.gca()
ax.spines['top'].set_color('none')
ax.xaxis.set_ticks_position('bottom')
ax.spines['bottom'].set_position(('data',0))
```

Me kan gjera det same med $y$-aksen.
Merk at me må definera heile plottet i ein boks i jupyter. Me kan ikkje modifisera eit eksisterande plott.

```python
plt.plot(x,y)
ax = plt.gca()
ax.spines['top'].set_color('none')
ax.xaxis.set_ticks_position('bottom')
ax.spines['bottom'].set_position(('data',0))
ax.spines['right'].set_color('none')
ax.yaxis.set_ticks_position('left')
ax.spines['left'].set_position(('data',0))

```
