
# Intro til jupyter notebook og jupyter lab


## Dokumentet - Hva er notebook?

- Hele nettleservinduet er jupyter lab = programmet / arbeidsmiljøet
- Selve dokumentet vi har åpnet er en jupyter notebook, `*.ipynb`
- Det finnes også en app som heter jupyter notebook som startet det hele -- litt forvirring av og til
- **Python** kjører i en *kernel* som lever separat fra selve dokumentet $\Rightarrow$ I terminalen

> Notebooken er arbeidsboka, kernel er kalkulatoren gjør utregningene og husker på utregninger du har gjort tidligere, jupyter lab er appen vi har åpnet arbeidsboka (notebook) med


Sjekk at du kan:
- Finne filområdet (venstre)
- Lage nye faner og lukke gamle
- åpne en ny notebook
- se hvilken *kernel* som brukes
- lagre en notebook

## Celler og navigasjon

Notebooks består av både tekstceller som denne, og kodeceller som cellen under

```python
print("Hei alle sammen")
```

- Vi kjører en celle ved å trykke på playknappen øverst, eller bruke `shift+enter` eller `ctrl+enter`
- Se om du ser forskjellen på de

Det er nyttig å lære seg noen "keyboard shortcuts", trykk `ctrl+shift+H` eller `shift-command-H` for å vise en liste

- Man kan skifte mellom kode- og text-celle med henholdsvis `Y` og `M`
- Man endrer en celle vå å gå i «edit» mode: trykke `enter`
- Man går ut av edit mode ved å trykke `Esc` 

I edit mode kan man bevege seg mellom celler med `J` og `K`

```python
# Dette er en kodecelle
10*2/5
```

```python
x=1
```

```python
x=2
```

```python
x=3
```
